
###  **Project Overview**

#### [**Problems Identified**]
#### **The Financial Lifecycle Crisis in the Indonesian Economy**
Based on the latest Consumer Expectation Index (IEK) from Bank Indonesia (March 2026), there’s a drop of 4.1 basis points since January. Even though it's still above 100, this is a signal that inflation is starting to hit hard, making average household spend increase every month.

Actually, this isn't just a post-pandemic thing. Looking back at 2019 data, the decline in purchasing power was already there. People became more cautious, but since living costs kept rising, many felt stuck. Some ended up picking investment instruments they didn't even understand, while others just stopped investing entirely because they were scared of the risk. The problem is, if they just keep their money in a regular savings account, it’s definitely going to lose to inflation.

So with our current economy, here are the core problems:
1. **Wedding Fund Struggle (Age 20s)**:

People in their 20s planning to get married need a huge amount of money. If they just save in a bank, the value will be eroded by inflation. But if they put it in stocks, there’s this constant fear: "What if the market crashes right when I need to pay the wedding vendors?" It’s a huge risk that could mess up their household finance from day one.

2. **KPR & Hidden Cost Overbudget (Age 30s)**:

By age 30, when people decide to buy a house and prepare for a Down Payment (DP), many realize it's way over their budget. Most people only focus on the DP and forget the Hidden Costs like Notary fees, BPHTB, and Bank Provisions (usually 0.5% - 3.5% of the loan). This lack of planning leads to households taking way longer KPR tenures because the financing doesn't match their current income.

3. **The Education Inflation Gap**:

After marriage, everyone wants the best education for their kids 10-18 years from now. But it's hard to imagine how much it will actually cost if inflation keeps weakening the Rupiah. People really need a tool to project these numbers accurately so they don't under-invest.

4. **Black Swan & Opportunity Cost**:

The 2020 crisis showed us that years of savings can be wiped out or stagnant in a flash. This creates a dilemma: stay in cash and lose to inflation, or go to the market and risk a crash. This Opportunity Cost is a major issue because people don't know how to build a portfolio that is actually "Black Swan resilient."

Based on the findings from the 2019–2025 dataset, the LQ45 composition underwent six routine evaluations occurring every February and August. Notably, in February 2024, stocks like INDY, SCMA, TBIG, and TPIA were excluded, while MBMA, MTEL, PGEO, and PTMP were added.
```python
# 1. Check Numerical Features Distribution (Line Charts (Trend), Histogram, Boxplot)
def plot_all_numerical(index):
    stock = emiten_clean[index]
    df = load_stock(stock)

    # Get list of numerical columns, drop columns with all NaN or constant values
    cols = df.select_dtypes(include=['number']).columns

    # Filter columns that have at least one non-NaN value
    cols = [col for col in cols if df[col].notna().any()]

    # Filter columns that have more than 1 unique value (not constant)
    cols = [col for col in cols if df[col].nunique() > 1]

    n_cols = len(cols)

    if n_cols == 0:
        print(f"No valid numerical columns for {stock}")
        return

    # Layout: 1 Numerical Column = 3 Plots (Trend, Hist, Box)
    fig, axes = plt.subplots(n_cols, 3, figsize=(32, 5 * n_cols))

    # Handle case when only one column exists
    if n_cols == 1:
        axes = axes.reshape(1, -1)

    for i, col in enumerate(cols):
        # Drop NaN for current column
        col_data = df[col].dropna()

        if len(col_data) == 0:
            continue

        # --- 1. TREND (TIME SERIES) ---
        axes[i, 0].plot(df.index, df[col], color='teal', linewidth=1)
        axes[i, 0].set_title(f'{col} - Trend Over Time', fontsize=12)
        axes[i, 0].set_xlabel('Date')
        axes[i, 0].set_ylabel('Value')
        axes[i, 0].tick_params(axis='x', rotation=30)
        axes[i, 0].grid(True, alpha=0.3)

        # --- 2. Distribution (Histogram) ---
        sns.histplot(col_data, kde=True, ax=axes[i, 1], color='purple', bins=50)
        axes[i, 1].set_title(f'{col} - Distribution', fontsize=12)
        axes[i, 1].set_xlabel('Value')
        axes[i, 1].set_ylabel('Frequency')

        # --- 3. Check Outliers (Boxplot) ---
        # Fix: Use DataFrame with explicit parameters
        sns.boxplot(x=col_data, ax=axes[i, 2], color='orange')
        axes[i, 2].set_title(f'{col} - Boxplot (Outlier Detection)', fontsize=12)
        axes[i, 2].set_xlabel('Value')

    plt.suptitle(f'Numerical Distribution: {stock}', fontsize=22, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()


# Interactive slider to check each stock
interact(plot_all_numerical,
         index=IntSlider(min=0, max=len(emiten_clean)-1, step=1,
                        continuous_update=False,
                        description='Stock Index:'))
```

```python
# Top Correlations
def show_top_correlations(index, top_n=10):
    stock = emiten_clean[index]
    df = load_stock(stock)

    # Get numerical columns
    num_cols = df.select_dtypes(include=['number']).dropna(axis=1, how='all')
    num_cols = num_cols.loc[:, num_cols.nunique() > 1]

    # Calculate correlation
    corr_matrix = num_cols.corr()

    # Extract upper triangle pairs
    pairs = []
    for i in range(len(corr_matrix.columns)):
        for j in range(i+1, len(corr_matrix.columns)):
            pairs.append({
                'Variable 1': corr_matrix.columns[i],
                'Variable 2': corr_matrix.columns[j],
                'Correlation': round(corr_matrix.iloc[i, j], 3)
            })

    corr_df = pd.DataFrame(pairs)

    # Separate positive and negative
    pos_df = corr_df[corr_df['Correlation'] > 0].sort_values('Correlation', ascending=False)
    neg_df = corr_df[corr_df['Correlation'] < 0].sort_values('Correlation', ascending=True)

    # Display
    print(f"\n{'='*60}")
    print(f"Correlation: Stock {stock}")
    print(f"{'='*60}")

    print(f"Top {top_n} Positive Correlations:")
    print("-"*50)
    for i in range(min(top_n, len(pos_df))):
        row = pos_df.iloc[i]
        print(f"{i+1:2}. {row['Variable 1']:15} ↔ {row['Variable 2']:15} : +{row['Correlation']:.3f}")

    print(f"Top {top_n} Negative Correlations:")
    print("-"*50)
    for i in range(min(top_n, len(neg_df))):
        row = neg_df.iloc[i]
        print(f"{i+1:2}. {row['Variable 1']:15} ↔ {row['Variable 2']:15} : {row['Correlation']:.3f}")


interact(show_top_correlations,
         index=IntSlider(min=0, max=len(emiten_clean)-1, step=1,
                        continuous_update=False,
                        description='Stock:'),
         top_n=IntSlider(min=5, max=30, step=5, value=10,
                        description='Top N:'))
```

```python
# 1. Daily and Rolling Returns
# Calculate daily returns and rolling period returns
# Daily return is the main input for boostrap sampling and risk calculation

# Create container for all processed data
all_stocks_data = {}

def calculate_returns(df):
    """
    Calculate daily and rolling returns for a stock.

    Parameters:
    df: DataFrame with 'close' column

    Returns:
    df with additional columns:
    - daily_return: (today's close - yesterday's close) / yesterday's close * 100
    - return_5d: return over 5 trading days
    - return_21d: return over ~1 month (21 trading days)
    - return_63d: return over ~3 months (63 trading days)
    """
    df = df.copy()

    # Daily return (%)
    df['daily_return'] = df['close'].pct_change() * 100

    # Rolling returns
    df['return_5d'] = df['close'].pct_change(5) * 100
    df['return_21d'] = df['close'].pct_change(21) * 100
    df['return_63d'] = df['close'].pct_change(63) * 100

    return df

# Apply and save to dictionary
for stock in emiten_clean:
    df = load_stock(stock)
    df_with_returns = calculate_returns(df)

    # Store in dictionary
    all_stocks_data[stock] = df_with_returns

# Verify
print(f"Total stocks processed: {len(all_stocks_data)}")
print(f"Columns added: daily_return, return_5d, return_21d, return_63d")

# Sample check
sample_stock = 'BBCA'
print(f"\nSample data for {sample_stock}:")
print(all_stocks_data[sample_stock][['close', 'daily_return', 'return_5d', 'return_21d']].tail(3))
```

```python
# 2. Volatility
# Because volatility measures how much a stock's price fluctuates. Higher volatility means higher risk.
# Volatility in this case gonna be used to classify stocks into risk categories for asset allocation recommendations

def calculate_volatility(df, window=20):
    """
    Calculate rolling volatility and risk classification.

    Parameters:
    df: DataFrame with 'daily_return' column
    window: rolling window in days (default: 20 days = 1 trading month)

    Returns:
    df with additional columns:
    - volatility_20d: rolling standard deviation of daily returns
    - volatility_annual: annualized volatility (× √252)
    - risk_level: Low / Medium / High classification
    """
    df = df.copy()

    # Rolling volatility
    df['volatility_20d'] = df['daily_return'].rolling(window=window).std()

    # Annualized volatility (252 trading days per year)
    df['volatility_annual'] = df['volatility_20d'] * (252 ** 0.5)

    # Risk classification
    # Low: < 1.5%, Medium: 1.5-3%, High: > 3%
    df['risk_level'] = pd.cut(
        df['volatility_20d'],
        bins=[0, 1.5, 3, 100],
        labels=['Low', 'Medium', 'High']
    )

    return df

# Apply to all stocks using dictionary
for stock in emiten_clean:
    df = all_stocks_data[stock]  # Get from dictionary
    df_with_volatility = calculate_volatility(df)
    all_stocks_data[stock] = df_with_volatility  # Update dictionary

# Sample check
print(f"Sample risk level for {sample_stock}:")
print(all_stocks_data[sample_stock][['daily_return', 'volatility_20d', 'risk_level']].tail(3))
```

```python
# 3. Drawdown
# Because drawdon measures the decline from a peak to trough
# Maximum Drawdown (MDD) represents the worst loss ever experienced
# Drawdown become essential for crisis simulation and answering "How bad can it get?"

def calculate_drawdown(df):
    """
    Calculate drawdown and maximum drawdown.

    Parameters:
    df: DataFrame with 'close' column

    Returns:
    df with additional columns:
    - cummax: cumulative maximum price
    - drawdown: percentage drop from peak (negative = loss)
    - max_drawdown: worst drawdown so far
    """
    df = df.copy()

    # Cumulative maximum price
    df['cummax'] = df['close'].cummax()

    # Drawdown (%) - negative values mean loss from peak
    df['drawdown'] = (df['close'] - df['cummax']) / df['cummax'] * 100

    # Maximum drawdown so far (most negative)
    df['max_drawdown'] = df['drawdown'].cummin()

    return df

# Apply to all stocks
for stock in emiten_clean:
    df = all_stocks_data[stock]
    df_with_drawdown = calculate_drawdown(df)
    all_stocks_data[stock] = df_with_drawdown

# Sample check
print(f"Sample drawdown for {sample_stock}:")
print(all_stocks_data[sample_stock][['close', 'drawdown', 'max_drawdown']].tail(3))
```

```python
# 4. VaR and Expected Shortfall
# This two metrics used for probability of success
# VaR (Value at Risk) is the maximum expected loss at a given confidence level (95%)
# Expected Shortfall (CVaR) is the average loss when losses exceed VaR

def calculate_var_es(returns, confidence_level=0.95):
    """
    Calculate Value at Risk (VaR) and Expected Shortfall (CVaR).

    Parameters:
    returns: array of daily returns (%)
    confidence_level: 0.95 for 95% VaR, 0.99 for 99% VaR

    Returns:
    VaR: Loss threshold at confidence_level
    CVaR: Average loss beyond VaR (tail risk)
    """
    returns = returns.dropna()

    # Historical VaR
    var = returns.quantile(1 - confidence_level)

    # Expected Shortfall (Conditional VaR)
    cvar = returns[returns <= var].mean()

    return var, cvar

# Store VaR results in a separate dataframe
var_results = []

for stock in emiten_clean:
    df = all_stocks_data[stock]
    returns = df['daily_return'].dropna()

    var_95, cvar_95 = calculate_var_es(returns, confidence_level=0.95)
    var_99, cvar_99 = calculate_var_es(returns, confidence_level=0.99)

    var_results.append({
        'Stock': stock,
        'VaR_95%': round(var_95, 2),
        'CVaR_95%': round(cvar_95, 2),
        'VaR_99%': round(var_99, 2),
        'CVaR_99%': round(cvar_99, 2)
    })

df_var = pd.DataFrame(var_results)
df_var = df_var.sort_values('VaR_95%', ascending=True)  # Lower VaR = less risky

print("Top 10 Least Risky Stocks (Lowest VaR_95%):")
print(df_var.head(10).to_string(index=False))

print("\nTop 10 Most Risky Stocks (Highest VaR_95%):")
print(df_var.tail(10).to_string(index=False))

all_stocks_data['_risk_metrics'] = df_var
```

```python
# 5. Crisis Flags
# This is used for weighted Bootstrap
# Creates flag for crisis periods and assign higher weights to those periods during Bootstrap sampling. This makes simulations more conservative

def add_crisis_flags(df):
    """
    Add crisis period flags and bootstrap weights.

    Parameters:
    df: DataFrame with datetime index

    Returns:
    df with additional columns:
    - is_crisis: True if date is during COVID crisis (Mar-Jun 2020)
    - bootstrap_weight: weight for resampling (3x for crisis period)
    """
    df = df.copy()

    # COVID-19 crisis period in Indonesia (March - June 2020)
    crisis_start = '2020-03-01'
    crisis_end = '2020-06-30'

    df['is_crisis'] = (df.index >= crisis_start) & (df.index <= crisis_end)

    # Bootstrap weights: crisis period gets 3x weight
    df['bootstrap_weight'] = 1.0
    df.loc[df['is_crisis'], 'bootstrap_weight'] = 3.0

    # Normalize weights to sum to 1
    df['bootstrap_weight'] = df['bootstrap_weight'] / df['bootstrap_weight'].sum()

    return df

# Apply to all stocks
for stock in emiten_clean:
    df = all_stocks_data[stock]
    df_with_flags = add_crisis_flags(df)
    all_stocks_data[stock] = df_with_flags

# Sample check
print(f"Crisis periods detected for {sample_stock}:")
print(f"Total crisis days: {all_stocks_data[sample_stock]['is_crisis'].sum()}")
print(f"Bootstrap weight sample: {all_stocks_data[sample_stock]['bootstrap_weight'].iloc[0]:.6f}")
```

```python
# 7. Inflation Adjusted Returns
# This is used for education fund
# Because the education inflation (5-15%) is higher than general inflation (2-5%)
# Consumer stocks historically can hedge against inflation. Real return shows actual purchasing power growth.

def calculate_real_return(nominal_return, inflation_rate):
    """Calculate real return after inflation."""
    real_return = ((1 + nominal_return/100) / (1 + inflation_rate/100)) - 1
    return real_return * 100

# Create empty list to store results
real_return_results = []

# Loop through all stocks
for stock in emiten_clean:
    df = all_stocks_data[stock]

    # Calculate average annual return
    avg_annual_return = df['daily_return'].mean() * 252

    # General inflation scenarios (2-5%)
    for scenario, inflation in [('low', 2.0), ('medium', 3.5), ('high', 5.0)]:
        real_return = calculate_real_return(avg_annual_return, inflation)
        real_return_results.append({
            'Stock': stock,
            'Nominal_Return_%': round(avg_annual_return, 2),
            'Scenario': scenario,
            'Inflation_%': inflation,
            'Real_Return_%': round(real_return, 2)
        })

    # Education inflation scenarios (5-15%)
    for scenario, inflation in [('edu_low', 5.0), ('edu_medium', 10.0), ('edu_high', 15.0)]:
        real_return = calculate_real_return(avg_annual_return, inflation)
        real_return_results.append({
            'Stock': stock,
            'Nominal_Return_%': round(avg_annual_return, 2),
            'Scenario': scenario,
            'Inflation_%': inflation,
            'Real_Return_%': round(real_return, 2)
        })

# Convert to DataFrame
df_real_returns = pd.DataFrame(real_return_results)

# INTERACTIVE INFLATION ANALYSIS
def run_inflation_analysis_interactive(stock, initial_capital, annual_inflation_pct):
    """
    Interactive analysis of real purchasing power growth.
    """
    df = all_stocks_data[stock]
    # Calculate historical average annual return (Mean Daily * 252 days)
    avg_annual_return = df['daily_return'].mean() * 252

    # Calculate Real Return based on User Input
    real_return = calculate_real_return(avg_annual_return, annual_inflation_pct)

    # Calculate Value Impact after 1 Year
    nominal_value_1y = initial_capital * (1 + avg_annual_return/100)
    real_value_1y = initial_capital * (1 + real_return/100)
    purchasing_power_loss = nominal_value_1y - real_value_1y

    # Visualization
    plt.figure(figsize=(12, 6))
    categories = ['Historical Nominal Return', f'Real Return (Adjusted for {annual_inflation_pct}% Inflation)']
    values = [avg_annual_return, real_return]
    colors = ['#34495e', '#e67e22' if real_return > 0 else '#c0392b']

    bars = plt.bar(categories, values, color=colors, alpha=0.8, width=0.6)
    plt.axhline(0, color='black', linewidth=0.8)

    # Add data labels
    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, yval, f'{yval:.2f}%',
                 ha='center', va='bottom' if yval > 0 else 'top', fontweight='bold')

    plt.title(f'Purchasing Power Analysis: {stock}', fontsize=12, fontweight='bold')
    plt.ylabel('Annual Return (%)')
    plt.grid(axis='y', alpha=0.3)

    plt.tight_layout()
    plt.show()

    # Terminal Output
    print("-" * 65)
    print(f"INFLATION ANALYSIS REPORT: {stock}")
    print("-" * 65)
    print(f"Selected Inflation Rate: {annual_inflation_pct}%")
    print(f"Historical Nominal Return: {avg_annual_return:.2f}%")
    print(f"Resulting Real Return: {real_return:.2f}%")
    print(f"\nMonetary Impact (1 Year):")
    print(f"  End Balance: Rp {nominal_value_1y:,.0f}")
    print(f"  Purchasing Power: Rp {real_value_1y:,.0f}")
    print(f"  Inflation Cost: Rp {purchasing_power_loss:,.0f}")

    if real_return < 0:
        print(f"\nADVISORY: Asset is failing to beat inflation. Purchasing power will decrease.")
    elif real_return < 5:
        print(f"\nADVISORY: Asset is a weak hedge. Barely maintaining purchasing power.")
    else:
        print(f"\nADVISORY: Strong inflation hedge. Real wealth is being created.")
    print("-" * 65)

# ============================================
# INTERACTIVE EXECUTION
# ============================================
interact(run_inflation_analysis_interactive,
         stock=Dropdown(options=emiten_clean, value='ICBP', description='Ticker:'),
         initial_capital=IntText(value=1_000_000, description='Capital (IDR):'),
         annual_inflation_pct=FloatSlider(min=2.0, max=15.0, step=0.5, value=10.0,
                                          description='Inflation %:',
                                          style={'description_width': 'initial'}))

# DISPLAY FULL TABLE (10% EDUCATION INFLATION)
print("\n" + "="*80)
print("FULL TABLE: REAL RETURN WITH 10% EDUCATION INFLATION")
print("="*80)
edu_table = df_real_returns[df_real_returns['Scenario'] == 'edu_medium'].copy()
edu_table = edu_table.sort_values('Real_Return_%', ascending=False)
print(edu_table[['Stock', 'Nominal_Return_%', 'Real_Return_%']].to_string(index=False))

print("\n" + "="*80)
print("TOP 10 HEDGE STOCKS FOR EDUCATION FUND")
print("="*80)
print(edu_table.head(10)[['Stock', 'Nominal_Return_%', 'Real_Return_%']].to_string(index=False))

print("\n" + "="*80)
print("WORST 10 STOCKS FOR EDUCATION FUND (Avoid)")
print("="*80)
print(edu_table.tail(10)[['Stock', 'Nominal_Return_%', 'Real_Return_%']].to_string(index=False))
```

```python
# 9. Bootstrap Function
# Bootstrap Function is bootstrap resamples historical returns (with crisis weighting) to simulate possible future returns

def bootstrap_simulation(returns, weights, n_simulations=10000, n_days=252):
    """
    Weighted bootstrap simulation for future returns.

    Parameters:
    returns: array of historical daily returns (%)
    weights: probability weights for each return
    n_simulations: number of simulation paths
    n_days: investment period in trading days

    Returns:
    cumulative_returns: array of total returns for each simulation (%)
    """
    # FIX: Ensure weights are float64 and sum exactly to 1.0 to prevent TypeError
    weights = np.array(weights).astype('float64')
    weights = weights / np.sum(weights)

    # Weighted sampling with replacement
    sampled_returns = np.random.choice(
        returns,
        size=(n_simulations, n_days),
        p=weights,
        replace=True
    )

    # Calculate cumulative return
    cumulative_returns = (1 + sampled_returns/100).prod(axis=1) - 1
    cumulative_returns = cumulative_returns * 100

    return cumulative_returns

def get_success_probability(target_return, simulated_returns):
    """Calculate probability of achieving target return."""
    success_count = np.sum(simulated_returns >= target_return)
    probability = (success_count / len(simulated_returns)) * 100
    return probability

def simulate_investment(stock, initial_investment, years, n_simulations=10000, use_crisis_weight=True):
    """
    Simulate investment return using bootstrap.

    Parameters:
    stock: stock ticker
    initial_investment: amount in IDR
    years: investment horizon in years
    n_simulations: number of simulations
    use_crisis_weight: if True, give more weight to COVID period

    Returns:
    final_values: array of simulated final values
    simulated_returns: array of simulated total returns (%)
    """
    df = all_stocks_data[stock]
    returns = df['daily_return'].dropna().values

    if use_crisis_weight:
        weights = df.loc[df['daily_return'].notna(), 'bootstrap_weight'].values if 'bootstrap_weight' in df.columns else np.ones(len(returns)) / len(returns)
    else:
        weights = np.ones(len(returns)) / len(returns)

    trading_days = int(years * 252)
    simulated_returns = bootstrap_simulation(returns, weights, n_simulations, trading_days)
    final_values = initial_investment * (1 + simulated_returns/100)

    return final_values, simulated_returns


def run_bootstrap_interactive(stock, initial_investment, target_fund, years, n_simulations, use_crisis_weight):
    """
    Strategic wrapper for the bootstrap engine with technical risk assessment.
    """
    # 1. Execute Simulation Engine
    final_values, sim_returns = simulate_investment(
        stock, initial_investment, years, n_simulations, use_crisis_weight
    )

    # 2. Key Performance Indicators (KPIs)
    success_count = np.sum(final_values >= target_fund)
    probability = (success_count / n_simulations) * 100

    # 3. Risk Metrics (95% Confidence Interval)
    var_95_value = np.percentile(final_values, 5)
    expected_shortfall = final_values[final_values <= var_95_value].mean()
    median_final_value = np.median(final_values)

    # 4. Data Visualization
    plt.figure(figsize=(15, 7))

    # Distribution of outcomes
    plt.hist(final_values, bins=100, color='#2c3e50', alpha=0.15, label='Scenario Distribution')
    plt.hist(final_values, bins=100, color='#34495e', histtype='step', linewidth=1)

    # Critical Thresholds
    plt.axvline(target_fund, color='#e74c3c', linestyle='--', linewidth=1.5, label=f'Target Threshold: Rp {target_fund:,.0f}')
    plt.axvline(median_final_value, color='#2980b9', linestyle='-', linewidth=2, label=f'Median Outcome: Rp {median_final_value:,.0f}')

    # Risk Area Visualization (Bottom 5%)
    plt.axvspan(final_values.min(), var_95_value, color='#c0392b', alpha=0.1, label='Critical Risk Zone (VaR 95%)')

    # Formatting
    plt.title(f'Monte Carlo Bootstrap Projection: {stock}', fontsize=12, fontweight='bold', loc='left')
    plt.xlabel('Projected Portfolio Value (IDR)', fontsize=10)
    plt.ylabel('Frequency of Scenarios', fontsize=10)
    plt.legend(loc='upper right', frameon=True, fontsize=9)
    plt.grid(True, alpha=0.15, linestyle='--')

    plt.tight_layout()
    plt.show()

    # 5. Technical Terminal Output
    print("-" * 65)
    print(f"SIMULATION REPORT: {stock}")
    print("-" * 65)
    print(f"Parameters: {years} Years | Target: Rp {target_fund:,.0f} | Crisis Weight: {use_crisis_weight}")
    print(f"Success Probability: {probability:.2f}%")
    print(f"\nQUANTITATIVE OUTLOOK:")
    print(f"  Median Case Projection:  Rp {median_final_value:,.0f}")
    print(f"  Stress Case Projection:  Rp {var_95_value:,.0f} (95% Confidence)")
    print(f"  Expected Shortfall:      Rp {expected_shortfall:,.0f}")

    if probability < 60:
        shortfall = target_fund - median_final_value
        print(f"\nADVISORY: Potential capital shortfall of Rp {max(0, shortfall):,.0f} detected.")
        print(f"STRATEGY: Adjust monthly contributions or reallocate to lower volatility assets.")
    else:
        print(f"\nADVISORY: Portfolio trajectory is aligned with historical benchmarks.")

    print(f"PLAN STATUS: {'VALIDATED' if probability > 75 else 'UNDER-CAPITALIZED'}")
    print("-" * 65)

# DASHBOARD EXECUTION
interact(run_bootstrap_interactive,
         stock=Dropdown(options=emiten_clean, value='BBCA', description='Ticker:'),
         initial_investment=IntText(value=1_000_000, description='Capital (IDR):', style={'description_width': 'initial'}),
         target_fund=IntText(value=150_000_000, description='Target (IDR):', style={'description_width': 'initial'}),
         years=FloatSlider(min=1, max=18, step=1, value=5, description='Horizon (Yrs):'),
         n_simulations=Dropdown(options=[1000, 5000, 10000], value=5000, description='Paths:'),
         use_crisis_weight=Checkbox(value=True, description='Apply Crisis Weight'))
```

```python
# 10. Recovery Time - Crisis Resilience Engine
# Calculates how many days a stocks needs to recover to pre-crisis prices.
# Focus: Crisis "Elasticity" for Life Milestone Planning.

def calculate_recovery_time(df, crisis_start='2020-02-01', crisis_end='2020-05-31'):
    """
    Calculate recovery time from COVID crisis with Resilience Scoring.
    """
    df = df.copy()

    # Validation: Skip stocks that were not listed yet during the 2020 crisis
    if df.index.min() > pd.to_datetime(crisis_start):
        return {'stock': None, 'recovered': 'N/A (New IPO)'}

    # 1. Benchmark: Price before the crash (Highest in 1 month pre-crisis)
    pre_crisis_data = df.loc[:crisis_start].tail(21)
    if pre_crisis_data.empty: return {'stock': None, 'recovered': 'No Data'}
    pre_crisis_price = pre_crisis_data['close'].max()

    # 2. Crisis Trough (The 'Bottom')
    crisis_df = df.loc[crisis_start:crisis_end]
    if crisis_df.empty: return {'stock': None, 'recovered': 'No Data'}
    trough_price = crisis_df['close'].min()
    trough_date = crisis_df['close'].idxmin()

    # 3. Recovery Velocity
    # We look for the first date AFTER the trough where price >= pre-crisis price
    post_trough_df = df.loc[trough_date:]
    recovery_df = post_trough_df[post_trough_df['close'] >= pre_crisis_price]

    max_drop_pct = ((trough_price - pre_crisis_price) / pre_crisis_price) * 100

    if len(recovery_df) > 0:
        recovery_date = recovery_df.index[0]
        recovery_days = (recovery_date - trough_date).days
        recovered = True
    else:
        recovery_days = (df.index[-1] - trough_date).days
        recovered = False

    # 4. Resilience Score (0-100) - Combined metric of depth and speed
    # A shallow drop and fast recovery = high score
    speed_factor = max(0, 100 - (recovery_days / 10)) if recovered else 0
    drop_factor = max(0, 100 + max_drop_pct)
    resilience_score = (speed_factor * 0.7) + (drop_factor * 0.3)

    return {
        'pre_crisis_price': round(pre_crisis_price, 2),
        'trough_price': round(trough_price, 2),
        'trough_date': trough_date.strftime('%Y-%m-%d'),
        'max_drop_%': round(max_drop_pct, 2),
        'recovery_days': recovery_days,
        'recovered': recovered,
        'resilience_score': round(resilience_score, 1),
        'risk_category': 'Bulletproof' if resilience_score > 75 else ('Average' if resilience_score > 40 else 'Fragile')
    }

# Calculate recovery time for all stocks
recovery_results = []

for stock in emiten_clean:
    df = all_stocks_data[stock]
    result = calculate_recovery_time(df)
    if result['recovered'] != 'N/A (New IPO)': # Keep only stocks existing in 2020
        result['stock'] = stock
        recovery_results.append(result)

df_recovery = pd.DataFrame(recovery_results)
# Sort by resilience (Higher is better)
df_recovery = df_recovery.sort_values('resilience_score', ascending=False)

def run_visual_recovery_analysis(df_recovery, filter_category='All'):
    """
    Elevates the recovery analysis results into an interactive visual dashboard.
    """
    # Filter data based on user input
    if filter_category == 'All':
        plot_df = df_recovery
    else:
        plot_df = df_recovery[df_recovery['risk_category'] == filter_category]

    plt.figure(figsize=(16, 9))

    # Standardize palette for financial risk
    risk_palette = {'Bulletproof': '#2ecc71', 'Average': '#f1c40f', 'Fragile': '#e74c3c'}

    # Plotting the Resilience Map
    sns.scatterplot(data=plot_df, x='max_drop_%', y='resilience_score',
                    hue='risk_category', size='resilience_score',
                    palette=risk_palette, sizes=(100, 500), alpha=0.6)

    # Add labels to ALL stocks in the plot
    for i in range(plot_df.shape[0]):
        plt.text(plot_df['max_drop_%'].iloc[i] + 0.4,
                 plot_df['resilience_score'].iloc[i],
                 plot_df['stock'].iloc[i],
                 fontsize=9, alpha=0.8)

    plt.axhline(y=50, color='gray', linestyle='--', alpha=0.4)
    plt.title(f'Recovery Resilience Analysis: 2020 Crisis', fontsize=14, fontweight='bold')
    plt.xlabel('Maximum Drawdown (%)', fontsize=11)
    plt.ylabel('Resilience Score (0 = Fragile, 100 = Bulletproof)', fontsize=11)
    plt.legend(title='Risk Classification', loc='lower left', frameon=True)
    plt.grid(True, alpha=0.15)

    plt.tight_layout()
    plt.show()

# Interactive widget
interact(run_visual_recovery_analysis,
         df_recovery=fixed(df_recovery),
         filter_category=Dropdown(options=['All', 'Bulletproof', 'Average', 'Fragile'],
                                  value='All',
                                  description='Risk Group:'))

# ---------------------------------------------------------
print("Top stocks that bounced back strongest from the 2020 Black Swan event.")
print("Fastest & Most Resilient Stocks (Top 10):")
fastest = df_recovery[df_recovery['recovered'] == True].head(10)
# Added resilience_score and category for better insight
print(fastest[['stock', 'max_drop_%', 'recovery_days', 'resilience_score', 'risk_category']].to_string(index=False))

print("\nSlowest or Fragile Stocks (Bottom 10):")
# Showing stocks that struggle to bounce back
slowest = df_recovery.tail(10)
print(slowest[['stock', 'max_drop_%', 'recovery_days', 'recovered', 'risk_category']].to_string(index=False))

# Store results
all_stocks_data['_recovery_analysis'] = df_recovery
```

```python
# 11 BEI Trading Calender Integration
def add_bei_calendar(df):
    """
    Add BEI (Bursa Efek Indonesia) trading calendar columns.
    BEI operates Monday to Friday only.
    """
    df = df.copy()

    # Basic trading calendar
    df['day_of_week'] = df.index.dayofweek
    df['day_name'] = df.index.day_name()
    df['is_trading_day'] = df['day_of_week'] <= 4  # Monday=0 to Friday=4

    # Month-end and quarter-end detection
    df['year_month'] = df.index.to_period('M')
    df['is_month_end'] = False
    df['year_quarter'] = df.index.to_period('Q')
    df['is_quarter_end'] = False

    for period in df['year_month'].unique():
        month_data = df[df['year_month'] == period]
        if len(month_data) > 0:
            last_day = month_data[month_data['is_trading_day']].index[-1]
            df.loc[last_day, 'is_month_end'] = True

    for period in df['year_quarter'].unique():
        quarter_data = df[df['year_quarter'] == period]
        if len(quarter_data) > 0:
            last_day = quarter_data[quarter_data['is_trading_day']].index[-1]
            df.loc[last_day, 'is_quarter_end'] = True

    # Drop temporary columns
    df = df.drop(['year_month', 'year_quarter'], axis=1)
    return df

def filter_trading_days(df):
    """Filter to only BEI trading days (Monday-Friday)."""
    df = add_bei_calendar(df)
    return df[df['is_trading_day']].copy()


# Apply to all stocks
for stock in emiten_clean:
    df = all_stocks_data[stock]
    df = filter_trading_days(df)
    all_stocks_data[stock] = df
```

```python
# 12. Moving Averages
# Moving averages identify the broader trend by smoothing out daily noise.
# In a financial plan, MA helps determine if it's a "Safe" time to accumulate
# or if you should "Protect" your capital during a downtrend.

# Common periods: MA20 (1 month), MA50 (2.5 months), MA200 (10 months)
# Standard Trading Signals:
# - Price above MA20 = Short-term bullish momentum
# - Price below MA20 = Short-term bearish momentum
# - MA20 above MA50 = Golden Cross (bullish trend confirmation)
# - MA20 below MA50 = Death Cross (bearish trend confirmation)

from ipywidgets import IntSlider, interact, IntText

def plot_moving_averages_strategic(index, short_ma=20, long_ma=50, investment_amount=1_000_000):
    """
    Strategic Trend Analysis with Capital Protection Insights.
    """
    stock = emiten_clean[index]
    df = load_stock(stock)

    # 1. Feature Engineering: Moving Averages
    df['MA_short'] = df['close'].rolling(window=short_ma).mean()
    df['MA_long'] = df['close'].rolling(window=long_ma).mean()

    # 2. Extract Current Values
    current_price = df['close'].iloc[-1]
    current_short = df['MA_short'].iloc[-1]
    current_long = df['MA_long'].iloc[-1]

    # Calculate previous values for crossover detection
    prev_short = df['MA_short'].iloc[-2] if len(df) > 1 else current_short
    prev_long = df['MA_long'].iloc[-2] if len(df) > 1 else current_long

    # 3. Dynamic Signal & Logic
    golden_cross = (prev_short <= prev_long) and (current_short > current_long)
    death_cross = (prev_short >= prev_long) and (current_short < current_long)

    if golden_cross:
        status, color, action = "BULLISH REVERSAL", "#2ecc71", "Begin Accumulation (Buy)"
        desc = "Golden Cross detected. Momentum is shifting upward."
    elif death_cross:
        status, color, action = "BEARISH REVERSAL", "#e74c3c", "Protect Capital (Reduce/Exit)"
        desc = "Death Cross detected. Significant risk of further decline."
    elif current_price > current_short and current_short > current_long:
        status, color, action = "STRONG UPTREND", "#27ae60", "Maintain Portfolio (Hold)"
        desc = "Price is well-supported by both short and long-term averages."
    elif current_price < current_short and current_short < current_long:
        status, color, action = "STRONG DOWNTREND", "#c0392b", "Wait for Stabilization (Cash)"
        desc = "Asset is underperforming. High risk for new entries."
    else:
        status, color, action = "Neutral / TRANSITION", "#f1c40f", "Monitor Closely (Watchlist)"
        desc = "Price is crossing averages. Trend is currently unclear."

    # 4. Investment Calculation
    dist_long_pct = ((current_price / current_long) - 1) * 100
    monetary_gap = investment_amount * (dist_long_pct / 100)
    valuation_status = "Premium" if dist_long_pct > 0 else "Discount"

    # 5. Visualization
    plt.figure(figsize=(15, 7))
    plt.plot(df.index, df['close'], label='Market Price', color='black', alpha=0.4, linewidth=1)
    plt.plot(df.index, df['MA_short'], label=f'Short-term (MA{short_ma})', color='#3498db', linewidth=2)
    plt.plot(df.index, df['MA_long'], label=f'Long-term (MA{long_ma})', color='#e67e22', linewidth=2)

    # Visualizing Trend Zones
    plt.fill_between(df.index, df['MA_short'], df['MA_long'],
                     where=(df['MA_short'] >= df['MA_long']),
                     color='#2ecc71', alpha=0.1, label='Bullish Momentum')
    plt.fill_between(df.index, df['MA_short'], df['MA_long'],
                     where=(df['MA_short'] < df['MA_long']),
                     color='#e74c3c', alpha=0.1, label='Bearish Momentum')

    plt.title(f'STRATEGIC TREND ANALYSIS: {stock} | {status}', fontsize=14, fontweight='bold')
    plt.ylabel('Price (IDR)')
    plt.legend(loc='lower right', fontsize=9)
    plt.grid(True, alpha=0.2, linestyle='--')

    plt.tight_layout()
    plt.show()

    # 6. Professional Summary Output
    print(f"\n{'='*65}")
    print(f"STRATEGIC POSITIONING SUMMARY: {stock}")
    print(f"{'='*65}")
    print(f"{'Current Market Price:':<25} {current_price:>15,.0f}")
    print(f"{f'MA{long_ma} Baseline:':<25} {current_long:>15,.0f}")
    print(f"{'Portfolio Valuation Gap:':<25} {dist_long_pct:>14.2f}%")
    print(f"\n{'Investment Amount:':<25} Rp {investment_amount:>12,.0f}")

    if dist_long_pct < 0:
        print(f"{'Market Discount Value:':<25} Rp {abs(monetary_gap):>12,.0f} (Potential Entry Opportunity)")
    else:
        print(f"{'Market Premium Value:':<25} Rp {monetary_gap:>12,.0f} (Potential Profit/Overvaluation)")

    print(f"\n{'Strategic Action:':<25} {action}")
    print(f"{'Trend Context:':<25} {desc}")
    print(f"{'='*65}")

# Interactive Widget
interact(plot_moving_averages_strategic,
         index=IntSlider(min=0, max=len(emiten_clean)-1, step=1,
                        continuous_update=False, description='Select Stock:',
                        style={'description_width': 'initial'}),
         short_ma=IntSlider(min=5, max=50, step=5, value=20,
                           description='Short Period:', style={'description_width': 'initial'}),
         long_ma=IntSlider(min=20, max=200, step=10, value=50,
                          description='Long Period:', style={'description_width': 'initial'}),
         investment_amount=IntText(value=1_000_000,
                                   description='Capital (Rp):', style={'description_width': 'initial'}))
```

```python
# 13. Rolling volatility
# Rolling volatility measures price fluctuation over a rolling window
# Higher volatility = higher risk = higher potential return

# Interpretation that we can get:
# - Low volatility (<1.5%): Stable stocks, suitable for short-term goals
# - Medium volatility (1.5-3%): Balanced risk-return
# - High volatility (>3%): Risky but high potential, for long-term goals
#
# Uses in financial planning:
# - Risk profiling (Conservative vs Aggressive)
# - Position sizing (reduce allocation on high volatility)
# - Crisis detection (spike in volatility signals market stress)

from ipywidgets import IntSlider, interact, IntText

def plot_rolling_volatility_professional(index, window=21, investment_amount=1_000_000):
    """
    Professional volatility analysis for life-stage financial planning.
    """
    stock = emiten_clean[index]
    df = load_stock(stock)

    # 1. Feature Engineering
    df['daily_return'] = df['close'].pct_change() * 100
    df['volatility'] = df['daily_return'].rolling(window=window).std()

    # 2. Metric Calculations
    current_vol = df['volatility'].dropna().iloc[-1]
    current_annual_vol = current_vol * np.sqrt(252) # Annualized Std Dev
    max_vol = df['volatility'].max()
    vol_95th = df['volatility'].quantile(0.95) # Panic Threshold

    # 3. Risk Classification Logic
    if current_vol <= 1.2:
        risk_status, risk_level, risk_color = "Stable/Conservative", "Low Risk", "#2ecc71"
        insight = "High predictability. Ideal for Wedding (1-3 yrs) or KPR (3-5 yrs) targets."
    elif current_vol <= 2.5:
        risk_status, risk_level, risk_color = "Moderate/Balanced", "Medium Risk", "#f1c40f"
        insight = "Standard equity risk. Suitable for 5-10 year horizons."
    else:
        risk_status, risk_level, risk_color = "Aggresive/Volatile", "High Risk", "#e74c3c"
        insight = "High market sentiment. Reserved for Child Education (10+ yrs)."

    # 4. Monetary Impact Analysis (Rupiah Impact)
    daily_impact_rupiah = investment_amount * (current_vol / 100)
    crisis_impact_rupiah = investment_amount * (max_vol / 100)

    # 5. Visualization
    plt.figure(figsize=(15, 7))
    plt.plot(df.index, df['volatility'], color='orange', linewidth=1.5,
             label=f'{window}-day Rolling Volatility', zorder=3)
    plt.fill_between(df.index, 0, df['volatility'], alpha=0.1, color='orange')

    # Crisis Stress Test Annotation (COVID-19 Period)
    covid_start, covid_end = pd.to_datetime('2020-03-01'), pd.to_datetime('2020-06-01')
    if covid_start in df.index:
        plt.axvspan(covid_start, covid_end, color='gray', alpha=0.15, label='COVID-19 Stress Period')
        crisis_peak = df.loc[covid_start:covid_end]['volatility'].max()
        crisis_peak_date = df.loc[covid_start:covid_end]['volatility'].idxmax()
        plt.annotate('Black Swan Event', xy=(crisis_peak_date, crisis_peak),
                     xytext=(crisis_peak_date, crisis_peak + 1.2),
                     ha='center', fontsize=9, fontweight='semibold', color='darkred',
                     arrowprops=dict(arrowstyle='->', color='darkred', lw=1))

    # Industry Reference Thresholds
    plt.axhline(y=1.5, color='#2ecc71', linestyle='--', alpha=0.5, label='Conservative Cap (1.5%)')
    plt.axhline(y=3.0, color='#e74c3c', linestyle='--', alpha=0.5, label='Aggressive Threshold (3.0%)')
    plt.axhline(y=vol_95th, color='purple', linestyle=':', alpha=0.7, label=f'Panic Level ({vol_95th:.2f}%)')

    plt.title(f'RISK PROFILE ANALYSIS: {stock} | {risk_status}', fontsize=14, fontweight='bold')
    plt.ylabel('Daily Volatility Magnitude (%)')
    plt.xlabel('Investment Period')
    plt.legend(loc='upper left', frameon=True, fontsize=9)
    plt.grid(True, alpha=0.2, linestyle='--')

    plt.tight_layout()
    plt.show()

    # Terminal Summary
    print(f"\n{'='*65}")
    print(f"STRATEGIC RISK SUMMARY: {stock}")
    print(f"{'='*65}")
    print(f"{'Current Volatility:':<25} {current_vol:.2f}% (Daily) | {current_annual_vol:.1f}% (Annualized)")
    print(f"{'Investment Basis:':<25} Rp {investment_amount:,.0f}")
    print(f"{'Normal Daily Fluctuation:':<25} ±Rp {daily_impact_rupiah:,.0f}")
    print(f"{'Extreme Crisis Fluctuation:':<25} ±Rp {crisis_impact_rupiah:,.0f}")
    print(f"\n{'Recommended Action:':<25} {insight}")
    print(f"{'='*65}")

# Interactive Dashboard Implementation
interact(plot_rolling_volatility_professional,
         index=IntSlider(min=0, max=len(emiten_clean)-1, step=1,
                        continuous_update=False, description='Select Stock:',
                        style={'description_width': 'initial'}),
         window=IntSlider(min=10, max=60, step=5, value=21,
                         description='Window (Days):',
                         style={'description_width': 'initial'}),
         investment_amount=IntText(value=1_000_000,
                                   description='Capital (Rp):',
                                   style={'description_width': 'initial'}))
```

```python
# ============================================
# 2. HISTORICAL RISK: DRAWDOWN (STRESS TEST VIEW)
# ============================================
# Focus: Shows deepest decline from peak (Peak-to-Trough)
# Critical for explaining Black Swan risk to users
# User can input investment amount to see potential loss in Rupiah

from ipywidgets import interact, IntSlider, IntText

def plot_drawdown_analysis(index, investment_amount=100_000_000):
    """
    Drawdown analysis with user investment impact.

    Parameters:
    index: stock index in emiten_clean list
    investment_amount: user's investment amount in IDR
    """
    stock = emiten_clean[index]
    df = load_stock(stock)

    # Calculate drawdown
    rolling_max = df['close'].cummax()
    drawdown = (df['close'] - rolling_max) / rolling_max * 100

    # Get current drawdown and maximum drawdown
    current_drawdown = drawdown.iloc[-1]
    max_drawdown = drawdown.min()
    max_drawdown_date = drawdown.idxmin()

    # Calculate potential losses in Rupiah
    current_loss = investment_amount * (abs(current_drawdown) / 100)
    max_loss = investment_amount * (abs(max_drawdown) / 100)

    plt.figure(figsize=(15, 6))
    plt.plot(df.index, drawdown, color='#d62728', linewidth=1.2)
    plt.fill_between(df.index, drawdown, 0, alpha=0.2, color='#d62728')

    # Annotate maximum drawdown
    plt.annotate(f'Max Drawdown: {max_drawdown:.1f}%',
                 xy=(max_drawdown_date, max_drawdown),
                 xytext=(max_drawdown_date, max_drawdown - 15),
                 arrowprops=dict(facecolor='black', shrink=0.05, width=1),
                 fontsize=10, fontweight='bold', color='darkred')

    # Annotate current drawdown
    current_date = drawdown.index[-1]
    plt.annotate(f'Current: {current_drawdown:.1f}%',
                 xy=(current_date, current_drawdown),
                 xytext=(current_date, current_drawdown - 8),
                 arrowprops=dict(facecolor='blue', shrink=0.05, width=0.8),
                 fontsize=9, color='blue')

    plt.title(f'Risk Profile: {stock} - Peak to Trough Decline (%)', fontsize=14, fontweight='bold')
    plt.axhline(-20, color='black', linestyle='--', alpha=0.5, label='Bear Market Threshold (-20%)')
    plt.ylabel('Drawdown (%)')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    # Print summary
    print(f"\n{'='*65}")
    print(f"DRAWDOWN ANALYSIS: {stock}")
    print(f"{'='*65}")
    print(f"Current Drawdown: {current_drawdown:.1f}%")
    print(f"  Paper Loss: Rp {current_loss:,.0f}")
    print(f"\nHistorical Max Drawdown: {max_drawdown:.1f}%")
    print(f"  Date: {max_drawdown_date.strftime('%Y-%m-%d')}")
    print(f"  Worst Loss: Rp {max_loss:,.0f}")
    print(f"\nRisk Warning:")
    print(f"  If you invest Rp {investment_amount:,.0f}, be prepared")
    print(f"  to potentially lose up to Rp {max_loss:,.0f} during market crashes.")

    # Risk classification based on max drawdown
    if abs(max_drawdown) < 20:
        print(f"\nRisk Level: LOW - Drawdown less than 20%")
    elif abs(max_drawdown) < 35:
        print(f"\nRisk Level: MODERATE - Drawdown 20-35%")
    else:
        print(f"\nRisk Level: HIGH - Drawdown over 35%")
    print(f"{'='*65}")

# Interactive widget
interact(plot_drawdown_analysis,
         index=IntSlider(min=0, max=len(emiten_clean)-1, step=1,
                        continuous_update=False,
                        description='Stock:',
                        style={'description_width': 'initial'}),
         investment_amount=IntText(value=1_000_000,
                                   description='Investment (Rp):',
                                   style={'description_width': 'initial'}))
```

```python
# ============================================
```
```python
# 6.2 KPR READINESS SIMULATOR
# ============================================

from ipywidgets import interact, IntText, FloatSlider, Dropdown, DatePicker, Output
from IPython.display import display, clear_output
from datetime import date, timedelta
import pandas as pd
import numpy as np

# Create output widget
kpr_output = Output()

# Helper functions (define if not already defined)
def calculate_kpr_hidden_costs(house_price, dp_percent):
    """Calculate hidden costs for KPR."""
    dp_amount = house_price * dp_percent / 100
    bphtb = house_price * 0.05
    notary_fee = house_price * 0.01
    bank_provision = (house_price - dp_amount) * 0.01
    total_needed = dp_amount + bphtb + notary_fee + bank_provision

    return {
        'down_payment': dp_amount,
        'bphtb': bphtb,
        'notary_fee': notary_fee,
        'bank_provision': bank_provision,
        'total_needed': total_needed
    }

def calculate_monthly_saving(target_amount, years, annual_return_pct):
    """Calculate monthly saving needed to reach target."""
    months = years * 12
    monthly_rate = (annual_return_pct / 100) / 12
    if monthly_rate == 0:
        return target_amount / months
    monthly_saving = target_amount * monthly_rate / ((1 + monthly_rate) ** months - 1)
    return monthly_saving

def get_stocks_by_risk_category(n_low=5, n_medium=5, n_high=5):
    """Auto-detect stocks based on VaR_95% from df_var."""
    try:
        all_stocks = df_var['Stock'].tolist()
        low_risk_stocks = all_stocks[:n_low]
        high_risk_stocks = all_stocks[-n_high:]
        mid_start = len(all_stocks) // 2 - n_medium // 2
        mid_end = mid_start + n_medium
        medium_risk_stocks = all_stocks[mid_start:mid_end]
        return low_risk_stocks, medium_risk_stocks, high_risk_stocks
    except:
        # Fallback if df_var not available
        return ['BBCA', 'TLKM', 'ASII', 'BBRI', 'ICBP'], ['CPIN', 'INDF', 'UNVR', 'ADRO', 'ITMG'], ['GOTO', 'BRPT', 'MEDC']

def get_kpr_portfolio_return(start_date=None):
    """Calculate expected portfolio return for KPR (60/30/10 allocation)."""
    low_stocks, medium_stocks, high_stocks = get_stocks_by_risk_category(5, 5, 5)

    # Use fixed returns for simplicity
    low_return = 0.08
    medium_return = 0.12
    high_return = 0.18

    portfolio_return = (0.60 * low_return + 0.30 * medium_return + 0.10 * high_return) * 100
    return portfolio_return, low_stocks[:3], medium_stocks[:3], high_stocks[:3]

def simulate_investment(stock, initial_investment, years, n_simulations=5000, use_crisis_weight=True):
    """Simplified bootstrap simulation."""
    # Use normal distribution for simulation when real data not available
    np.random.seed(42)
    mean_return = 0.0003  # ~0.03% daily average
    std_return = 0.02     # ~2% daily volatility

    n_days = int(years * 252)
    simulated_returns = np.random.normal(mean_return, std_return, (n_simulations, n_days))
    cumulative_returns = (1 + simulated_returns).prod(axis=1) - 1
    final_values = initial_investment * (1 + cumulative_returns)

    return final_values, cumulative_returns * 100

def get_success_probability(target, final_values):
    """Calculate probability of success."""
    return (final_values >= target).mean() * 100

def run_dividend_kpr_analysis(stock, investment_amount, monthly_kpr):
    """Simplified dividend analysis."""
    print(f"\nDividend Analysis for {stock}:")
    print(f"  Investment: Rp {investment_amount:,.0f}")
    print(f"  Est. yearly dividend: Rp {investment_amount * 0.035:,.0f} (3.5%)")
    print(f"  KPR coverage: {(investment_amount * 0.035) / monthly_kpr:.1f} months/year")

# Default start date
default_start = date.today() - timedelta(days=3*365)

def run_kpr_simulator(house_price, dp_percent, monthly_kpr, years, monthly_saving, risk_profile, start_date):
    """
    KPR Readiness Simulator.
    """
    with kpr_output:
        clear_output(wait=True)

        # Calculate end date
        end_date = start_date + timedelta(days=int(years * 365.25))

        # Calculate hidden costs
        costs = calculate_kpr_hidden_costs(house_price, dp_percent)

        print("=" * 65)
        print("KPR READINESS SIMULATOR")
        print("=" * 65)
        print(f"\nHouse Price:          Rp {house_price:,.0f}")
        print(f"Down Payment:         {dp_percent:.0f}% (Rp {costs['down_payment']:,.0f})")
        print(f"Monthly KPR:          Rp {monthly_kpr:,.0f}")
        print(f"\nInvestment Period:    {start_date.strftime('%B %d, %Y')} to {end_date.strftime('%B %d, %Y')}")
        print(f"Time Horizon:         {years} year(s)")
        print(f"Monthly Saving:       Rp {monthly_saving:,.0f}")

        # Hidden costs breakdown
        print(f"\n{'='*65}")
        print("TOTAL DP + HIDDEN COSTS")
        print(f"{'='*65}")
        print(f"Down Payment (DP):    Rp {costs['down_payment']:>15,.0f}")
        print(f"BPHTB (5%):           Rp {costs['bphtb']:>15,.0f}")
        print(f"Notary Fee (1%):      Rp {costs['notary_fee']:>15,.0f}")
        print(f"Bank Provision (1%):  Rp {costs['bank_provision']:>15,.0f}")
        print(f"{'-'*65}")
        print(f"TOTAL NEEDED:         Rp {costs['total_needed']:>15,.0f}")

        # Get portfolio return and stocks
        portfolio_return, low_stocks, medium_stocks, high_stocks = get_kpr_portfolio_return(start_date)

        # Calculate required monthly saving
        required_saving = calculate_monthly_saving(costs['total_needed'], years, portfolio_return)

        print(f"\n{'='*65}")
        print("INVESTMENT ANALYSIS")
        print(f"{'='*65}")
        print(f"Expected Portfolio Return: {portfolio_return:.1f}% per year")
        print(f"Required Monthly Saving:   Rp {required_saving:,.0f}")

        if monthly_saving >= required_saving:
            print(f"  Your saving of Rp {monthly_saving:,.0f} is SUFFICIENT for DP")
        else:
            shortfall = required_saving - monthly_saving
            print(f"  Your saving of Rp {monthly_saving:,.0f} is INSUFFICIENT")
            print(f"  Need additional Rp {shortfall:,.0f}/month")

        # Bootstrap simulation
        proxy_stock = low_stocks[0] if low_stocks else 'BBCA'
        initial_investment = monthly_saving * 12 * years
        final_values, sim_returns = simulate_investment(
            proxy_stock, initial_investment, years, n_simulations=5000, use_crisis_weight=True
        )
        probability = get_success_probability(costs['total_needed'], final_values)

        print(f"\nProbability of reaching DP target: {probability:.1f}%")

        # KPR Readiness Score
        if probability >= 80:
            kpr_score = "READY"
            read_status = "Your DP fund is on track. Proceed with KPR application."
        elif probability >= 50:
            kpr_score = "CAUTION"
            read_status = "Consider increasing savings or extending timeline."
        else:
            kpr_score = "NOT READY"
            read_status = "Postpone KPR or seek lower house price / higher DP."

        print(f"\nKPR Readiness Score:  {kpr_score}")
        print(f"Status:               {read_status}")

        # Portfolio recommendation
        print(f"\n{'='*65}")
        print("RECOMMENDED PORTFOLIO (60/30/10 Strategy)")
        print(f"{'='*65}")
        print(f"  - Low Risk (60%):   {', '.join(low_stocks[:3])}")
        print(f"  - Medium Risk (30%): {', '.join(medium_stocks[:3])}")
        print(f"  - High Risk (10%):  {', '.join(high_stocks[:3])}")

        # Dividend synergy
        print(f"\n{'='*65}")
        print("DIVIDEND SYNERGY ANALYSIS")
        print(f"{'='*65}")
        print("Based on bank stocks in your portfolio:")

        bank_stocks = ['BBCA', 'BBRI', 'BMRI', 'BBNI']
        for stock in low_stocks[:3] + medium_stocks[:3]:
            if stock in bank_stocks:
                estimated_yield = 0.035
                yearly_dividend = monthly_saving * 12 * estimated_yield
                months_covered = yearly_dividend / monthly_kpr if monthly_kpr > 0 else 0
                print(f"\n  {stock}:")
                print(f"    Est. Dividend Yield: 3.5%")
                print(f"    Est. Yearly Dividend: Rp {yearly_dividend:,.0f}")
                print(f"    Covers {months_covered:.1f} months of KPR")

        print("\n" + "=" * 65)
        print("Note: Dividends from bank stocks can help cover monthly KPR.")
        print("=" * 65)

# Interactive widget
interact(run_kpr_simulator,
         house_price=IntText(value=500_000_000, description='House Price (Rp):',
                             style={'description_width': 'initial'}),
         dp_percent=FloatSlider(min=5, max=30, step=5, value=20, description='DP Percentage:',
                                style={'description_width': 'initial'}),
         monthly_kpr=IntText(value=5_000_000, description='Monthly KPR (Rp):',
                             style={'description_width': 'initial'}),
         years=FloatSlider(min=3, max=5, step=0.5, value=4, description='Years to Save DP:',
                          style={'description_width': 'initial'}),
         monthly_saving=IntText(value=2_000_000, description='Monthly Saving (Rp):',
                                style={'description_width': 'initial'}),
         risk_profile=Dropdown(options=['Conservative', 'Moderate', 'Aggressive'],
                               value='Moderate', description='Risk Profile:',
                               style={'description_width': 'initial'}),
         start_date=DatePicker(value=default_start, description='Start Date:',
                               style={'description_width': 'initial'}))

display(kpr_output)
```

```python
# ============================================
```
```python
# 6.4 TAX & TRANSACTION COST CALCULATOR
# ============================================
# Based on Indonesian tax regulations for stock investments

from ipywidgets import interact, IntText, FloatSlider, Dropdown, Output
from IPython.display import display, clear_output

tax_output = Output()

def calculate_transaction_cost(investment_amount, transaction_fee_pct=0.2):
    """
    Calculate transaction cost for buying stocks.

    Parameters:
    investment_amount: Amount to invest (IDR)
    transaction_fee_pct: Broker fee percentage (0.1% - 0.3%)

    Returns:
    transaction_fee, total_cost, shares_received (approx)
    """
    transaction_fee = investment_amount * (transaction_fee_pct / 100)
    total_cost = investment_amount + transaction_fee

    return {
        'investment_amount': investment_amount,
        'transaction_fee_pct': transaction_fee_pct,
        'transaction_fee': transaction_fee,
        'total_cost': total_cost
    }

def calculate_dividend_tax(dividend_received, dividend_tax_pct=10):
    """
    Calculate dividend tax.

    Parameters:
    dividend_received: Gross dividend received (IDR)
    dividend_tax_pct: Dividend tax percentage (default 10%)

    Returns:
    tax_amount, net_dividend
    """
    tax_amount = dividend_received * (dividend_tax_pct / 100)
    net_dividend = dividend_received - tax_amount

    return {
        'gross_dividend': dividend_received,
        'tax_pct': dividend_tax_pct,
        'tax_amount': tax_amount,
        'net_dividend': net_dividend
    }

def calculate_capital_gain_tax(buy_price, sell_price, shares, capital_gain_tax_pct=0.1):
    """
    Calculate capital gain tax.

    Parameters:
    buy_price: Price per share when buying (IDR)
    sell_price: Price per share when selling (IDR)
    shares: Number of shares
    capital_gain_tax_pct: Capital gain tax percentage (default 0.1%)

    Returns:
    capital_gain, tax_amount, net_profit
    """
    total_buy = buy_price * shares
    total_sell = sell_price * shares
    capital_gain = total_sell - total_buy

    if capital_gain > 0:
        tax_amount = capital_gain * (capital_gain_tax_pct / 100)
    else:
        tax_amount = 0

    net_profit = capital_gain - tax_amount

    return {
        'buy_price': buy_price,
        'sell_price': sell_price,
        'shares': shares,
        'total_buy': total_buy,
        'total_sell': total_sell,
        'capital_gain': capital_gain,
        'tax_pct': capital_gain_tax_pct,
        'tax_amount': tax_amount,
        'net_profit': net_profit
    }

def calculate_total_return_with_tax(initial_investment, final_value, dividend_received=0,
                                     transaction_fee_pct=0.2, dividend_tax_pct=10):
    """
    Calculate total return after all taxes and fees.
    """
    # Transaction cost when buying
    buy_cost = calculate_transaction_cost(initial_investment, transaction_fee_pct)
    actual_investment = buy_cost['total_cost']

    # Transaction cost when selling
    sell_cost = calculate_transaction_cost(final_value, transaction_fee_pct)
    gross_sell = final_value - sell_cost['transaction_fee']

    # Dividend tax
    if dividend_received > 0:
        dividend_tax = calculate_dividend_tax(dividend_received, dividend_tax_pct)
        net_dividend = dividend_tax['net_dividend']
    else:
        net_dividend = 0

    # Capital gain tax (only if positive)
    capital_gain = gross_sell - actual_investment
    if capital_gain > 0:
        cap_gain_tax = calculate_capital_gain_tax(
            actual_investment / (initial_investment / 1000),
            gross_sell / (initial_investment / 1000),
            initial_investment / 1000,
            0.1
        )
        cap_gain_tax_amount = cap_gain_tax['tax_amount']
    else:
        cap_gain_tax_amount = 0

    net_final = gross_sell + net_dividend - cap_gain_tax_amount
    total_return = ((net_final - initial_investment) / initial_investment) * 100

    return {
        'initial_investment': initial_investment,
        'final_value_before_tax': final_value,
        'transaction_fee_buy': buy_cost['transaction_fee'],
        'transaction_fee_sell': sell_cost['transaction_fee'],
        'gross_dividend': dividend_received,
        'net_dividend': net_dividend,
        'capital_gain_tax': cap_gain_tax_amount,
        'net_final_value': net_final,
        'total_return_pct': total_return
    }

def run_tax_calculator(initial_investment, expected_return_pct, years, dividend_yield_pct, transaction_fee_pct, dividend_tax_pct):
    """
    Interactive Tax & Transaction Cost Calculator.
    """
    with tax_output:
        clear_output(wait=True)

        # Calculate final value before tax
        final_value_before_tax = initial_investment * ((1 + expected_return_pct/100) ** years)

        # Calculate total dividend received
        annual_dividend = initial_investment * (dividend_yield_pct / 100)
        total_dividend = annual_dividend * years

        # Calculate all taxes and fees
        result = calculate_total_return_with_tax(
            initial_investment, final_value_before_tax, total_dividend,
            transaction_fee_pct, dividend_tax_pct
        )

        print("=" * 65)
        print("TAX & TRANSACTION COST CALCULATOR")
        print("=" * 65)
        print(f"\nInitial Investment:     Rp {initial_investment:,.0f}")
        print(f"Expected Return:        {expected_return_pct:.1f}% per year")
        print(f"Investment Period:      {years} years")
        print(f"Dividend Yield:         {dividend_yield_pct:.2f}% per year")
        print(f"Transaction Fee:        {transaction_fee_pct:.2f}%")
        print(f"Dividend Tax Rate:      {dividend_tax_pct:.0f}%")

        print(f"\n{'='*65}")
        print("COST BREAKDOWN")
        print(f"{'='*65}")
        print(f"Final Value (Before Tax): Rp {final_value_before_tax:,.0f}")
        print(f"Transaction Fee (Buy):    Rp {result['transaction_fee_buy']:,.0f}")
        print(f"Transaction Fee (Sell):   Rp {result['transaction_fee_sell']:,.0f}")
        print(f"Gross Dividend ({years} years): Rp {total_dividend:,.0f}")
        print(f"Dividend Tax (10%):        Rp {result['gross_dividend'] - result['net_dividend']:,.0f}")

        if result['capital_gain_tax'] > 0:
            print(f"Capital Gain Tax (0.1%):  Rp {result['capital_gain_tax']:,.0f}")

        print(f"\n{'='*65}")
        print("FINAL RESULT")
        print(f"{'='*65}")
        print(f"Net Final Value:          Rp {result['net_final_value']:,.0f}")
        print(f"Net Profit:               Rp {result['net_final_value'] - initial_investment:,.0f}")
        print(f"Total Return (After Tax): {result['total_return_pct']:.1f}%")

        # Tax efficiency ratio
        total_tax = (result['transaction_fee_buy'] + result['transaction_fee_sell'] +
                     (result['gross_dividend'] - result['net_dividend']) +
                     result['capital_gain_tax'])
        tax_efficiency = ((result['net_final_value'] - initial_investment) /
                         (final_value_before_tax - initial_investment)) * 100 if final_value_before_tax > initial_investment else 0

        print(f"\nTotal Tax & Fees:         Rp {total_tax:,.0f}")
        print(f"Tax Efficiency:           {tax_efficiency:.1f}% of gross profit retained")

        print("=" * 65)
        print("\nNotes:")
        print("  - Transaction fee: 0.1% - 0.3% depends on broker")
        print("  - Dividend tax: 10% final tax (withholding tax)")
        print("  - Capital gain tax: 0.1% for listed stocks (final tax)")
        print("=" * 65)


# Interactive Widget
interact(run_tax_calculator,
         initial_investment=IntText(value=100_000_000, description='Initial Investment (Rp):',
                                    style={'description_width': 'initial'}),
         expected_return_pct=FloatSlider(min=0, max=30, step=0.5, value=12,
                                        description='Expected Return (%/year):',
                                        style={'description_width': 'initial'}),
         years=IntSlider(min=1, max=20, step=1, value=5, description='Years:',
                        style={'description_width': 'initial'}),
         dividend_yield_pct=FloatSlider(min=0, max=10, step=0.1, value=2.5,
                                       description='Dividend Yield (%):',
                                       style={'description_width': 'initial'}),
         transaction_fee_pct=FloatSlider(min=0.1, max=0.3, step=0.01, value=0.2,
                                        description='Transaction Fee (%):',
                                        style={'description_width': 'initial'}),
         dividend_tax_pct=FloatSlider(min=10, max=10, step=0, value=10,
                                     description='Dividend Tax (%):',
                                     style={'description_width': 'initial'}))

display(tax_output)
```

```python
# ============================================
# 6.5 MONTE CARLO SIMULATION
# ============================================
# Monte Carlo simulation using bootstrap resampling
# Projects future portfolio values based on historical returns
# Outputs: 5th, 50th, 95th percentiles and probability of success

from ipywidgets import interact, IntText, FloatSlider, Dropdown, Output
from IPython.display import display, clear_output

monte_carlo_output = Output()

def run_monte_carlo_simulation(stock, initial_investment, years, target_amount,
                               n_simulations=10000, use_crisis_weight=True):
    """
    Monte Carlo Simulation for investment projection.

    Parameters:
    stock: Stock ticker
    initial_investment: Initial investment amount (IDR)
    years: Investment horizon (years)
    target_amount: Target amount to achieve (IDR)
    n_simulations: Number of simulation paths (default 10000)
    use_crisis_weight: Include COVID crisis weighting (default True)

    Outputs:
    - Distribution of final values
    - 5th, 50th, 95th percentiles
    - Probability of success
    - Value at Risk (VaR 95%)
    """
    with monte_carlo_output:
        clear_output(wait=True)

        # Run bootstrap simulation
        final_values, sim_returns = simulate_investment(
            stock, initial_investment, years, n_simulations, use_crisis_weight
        )

        # Calculate percentiles
        p5 = np.percentile(final_values, 5)
        p50 = np.percentile(final_values, 50)
        p95 = np.percentile(final_values, 95)

        # Calculate probability of success
        probability = get_success_probability(target_amount, final_values)

        # Calculate VaR (95% confidence)
        var_95 = np.percentile(final_values, 5)
        expected_shortfall = final_values[final_values <= var_95].mean()

        print("=" * 65)
        print("MONTE CARLO SIMULATION")
        print("=" * 65)
        print(f"\nStock:                  {stock}")
        print(f"Initial Investment:     Rp {initial_investment:,.0f}")
        print(f"Time Horizon:           {years} years")
        print(f"Target Amount:          Rp {target_amount:,.0f}")
        print(f"Simulation Paths:       {n_simulations:,}")
        print(f"Crisis Weighting:       {'Yes' if use_crisis_weight else 'No'}")

        print(f"\n{'='*65}")
        print("PROJECTION PERCENTILES")
        print(f"{'='*65}")
        print(f"  Pessimistic (5th):    Rp {p5:,.0f}")
        print(f"  Moderate (50th):      Rp {p50:,.0f}")
        print(f"  Optimistic (95th):    Rp {p95:,.0f}")

        print(f"\n{'='*65}")
        print("RISK METRICS")
        print(f"{'='*65}")
        print(f"Probability of Success: {probability:.1f}%")
        print(f"Value at Risk (VaR 95%): Rp {var_95:,.0f}")
        print(f"Expected Shortfall:     Rp {expected_shortfall:,.0f}")

        # Interpretation
        print(f"\n{'='*65}")
        print("INTERPRETATION")
        print(f"{'='*65}")

        if probability >= 80:
            print(f"  Status: ON TRACK")
            print(f"  Your investment has a high probability of reaching the target.")
        elif probability >= 50:
            print(f"  Status: CAUTION")
            print(f"  Only 1 in 2 scenarios reach the target. Consider increasing savings.")
        else:
            print(f"  Status: AT RISK")
            print(f"  More than half of scenarios miss the target.")

        print(f"\n  95% of outcomes fall between Rp {p5:,.0f} and Rp {p95:,.0f}")
        print(f"  There is a 5% chance of losing below Rp {var_95:,.0f}")

        print("=" * 65)

        # Visualization
        plt.figure(figsize=(14, 6))

        # Histogram of outcomes
        plt.hist(final_values, bins=100, color='steelblue', alpha=0.6, edgecolor='black', linewidth=0.5)
        plt.axvline(target_amount, color='red', linestyle='--', linewidth=2, label=f'Target: Rp {target_amount:,.0f}')
        plt.axvline(p50, color='green', linestyle='-', linewidth=2, label=f'Median: Rp {p50:,.0f}')
        plt.axvline(var_95, color='orange', linestyle=':', linewidth=2, label=f'VaR 95%: Rp {var_95:,.0f}')

        plt.title(f'Monte Carlo Simulation: {stock} ({years} years, {n_simulations:,} scenarios)',
                  fontsize=12, fontweight='bold')
        plt.xlabel('Final Portfolio Value (IDR)')
        plt.ylabel('Frequency')
        plt.legend(loc='upper right')
        plt.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()


# Interactive Widget
interact(run_monte_carlo_simulation,
         stock=Dropdown(options=emiten_clean, value='BBCA', description='Stock:',
                        style={'description_width': 'initial'}),
         initial_investment=IntText(value=100_000_000, description='Initial Investment (Rp):',
                                    style={'description_width': 'initial'}),
         years=FloatSlider(min=1, max=20, step=1, value=5, description='Years:',
                          style={'description_width': 'initial'}),
         target_amount=IntText(value=200_000_000, description='Target Amount (Rp):',
                               style={'description_width': 'initial'}),
         n_simulations=Dropdown(options=[1000, 5000, 10000, 50000], value=10000,
                               description='Simulations:',
                               style={'description_width': 'initial'}),
         use_crisis_weight=Dropdown(options=[True, False], value=True,
                                   description='Crisis Weight:',
                                   style={'description_width': 'initial'}))

display(monte_carlo_output)
```

```python
# ============================================
# 6.6 RECOMMENDATION ENGINE
# ============================================
# Data-driven stock selection based on goal, risk profile, and historical performance
# Metrics used: annual return, volatility, maximum drawdown

import pandas as pd
import numpy as np
from typing import Dict


def safe_normalize(series: pd.Series) -> pd.Series:
    """Safe min-max normalization that handles constant series."""
    if series.max() == series.min():
        return pd.Series(0.5, index=series.index)
    return (series - series.min()) / (series.max() - series.min())


def classify_risk(volatility: float) -> str:
    """
    Classify stock risk based on annualized volatility (in percentage).
    Low: < 18%, Medium: 18-30%, High: > 30%
    """
    if volatility < 18.0:
        return "Low Risk"
    elif volatility < 30.0:
        return "Medium Risk"
    else:
        return "High Risk"


def generate_stock_insight(stock: str, df_features: pd.DataFrame) -> str:
    """Generate simple insight why a stock was selected."""
    row = df_features.loc[stock]
    insights = []

    if row["annual_return"] > 12.0:
        insights.append("high return")
    if row["volatility"] < 20.0:
        insights.append("low volatility")
    if row["max_drawdown"] * 100 > -30.0:
        insights.append("crisis resilient")

    if not insights:
        return f"{stock}: balanced metrics"

    return f"{stock}: {' + '.join(insights)}"


def compute_stock_features(df: pd.DataFrame) -> Dict[str, float]:
    """Extract key features from a stock's historical data."""
    df = df.copy()

    df["daily_ret"] = df["close"].pct_change()

    annual_return = df["daily_ret"].mean() * 252 * 100
    volatility = df["daily_ret"].std() * np.sqrt(252) * 100

    cummax = df["close"].cummax()
    drawdown = (df["close"] - cummax) / cummax * 100
    max_drawdown = drawdown.min() / 100

    return {
        "annual_return": annual_return,
        "volatility": volatility,
        "max_drawdown": max_drawdown
    }


def build_stock_features_dataframe(all_stocks_data: Dict[str, pd.DataFrame]) -> pd.DataFrame:
    """Build a dataframe with features for all stocks."""
    features_list = []

    for stock, df in all_stocks_data.items():
        if not isinstance(df, pd.DataFrame):
            continue
        if 'close' not in df.columns:
            continue

        try:
            features = compute_stock_features(df)
            features["stock"] = stock
            features_list.append(features)
        except Exception as e:
            print(f"Warning: Could not process {stock}: {e}")
            continue

    df_features = pd.DataFrame(features_list)
    df_features = df_features.set_index("stock")

    return df_features


def normalize_features(df_features: pd.DataFrame) -> pd.DataFrame:
    """Normalize features to 0-1 scale."""
    df_norm = df_features.copy()

    df_norm["return_norm"] = safe_normalize(df_features["annual_return"])
    df_norm["vol_norm"] = 1 - safe_normalize(df_features["volatility"])
    df_norm["drawdown_norm"] = 1 - safe_normalize(df_features["max_drawdown"])

    return df_norm


def get_goal_weights(goal: str) -> Dict[str, float]:
    """Return weights based on investment goal."""
    if "Wedding" in goal:
        return {"return": 0.30, "volatility": 0.45, "drawdown": 0.25}
    elif "KPR" in goal:
        return {"return": 0.45, "volatility": 0.35, "drawdown": 0.20}
    else:
        return {"return": 0.55, "volatility": 0.25, "drawdown": 0.20}


def adjust_weights_for_risk(weights: Dict[str, float], risk_profile: str) -> Dict[str, float]:
    """Adjust weights based on user's risk tolerance."""
    adjusted = weights.copy()

    if risk_profile == "Conservative":
        adjusted["volatility"] += 0.10
        adjusted["drawdown"] += 0.10
        adjusted["return"] -= 0.20
    elif risk_profile == "Aggressive":
        adjusted["return"] += 0.15
        adjusted["volatility"] -= 0.10
        adjusted["drawdown"] -= 0.05

    total = sum(adjusted.values())
    return {k: v / total for k, v in adjusted.items()}


def score_stocks(df_norm: pd.DataFrame, weights: Dict[str, float]) -> pd.DataFrame:
    """Calculate final score for each stock based on weights."""
    df_score = df_norm.copy()

    df_score["final_score"] = (
        weights["return"] * df_score["return_norm"] +
        weights["volatility"] * df_score["vol_norm"] +
        weights["drawdown"] * df_score["drawdown_norm"]
    )

    return df_score.sort_values("final_score", ascending=False)


def get_smooth_allocations(scores: pd.Series) -> Dict[str, float]:
    """Convert scores to allocations using softmax for smoother distribution."""
    std_val = scores.std()
    if std_val == 0:
        exp_scores = np.ones(len(scores))
    else:
        exp_scores = np.exp(scores / std_val)

    softmax_alloc = exp_scores / exp_scores.sum()

    allocations = {stock: round(alloc * 100, 1) for stock, alloc in zip(scores.index, softmax_alloc)}

    total = sum(allocations.values())
    if total != 100:
        first_stock = list(allocations.keys())[0]
        allocations[first_stock] += round(100 - total, 1)

    return allocations


def get_strategy_explanation(goal: str, risk_profile: str) -> str:
    """Generate strategy explanation based on goal and risk profile."""
    if "Wedding" in goal:
        base_strategy = "80% low risk + 20% medium risk stocks"
        focus = "capital protection"
    elif "KPR" in goal:
        base_strategy = "60% low risk + 30% medium risk + 10% high risk"
        focus = "balanced return with moderate risk"
    else:
        base_strategy = "50% low risk + 30% medium risk + 20% high risk"
        focus = "long-term growth"

    risk_note = ""
    if risk_profile == "Conservative":
        risk_note = " with extra weight on stability"
    elif risk_profile == "Aggressive":
        risk_note = " with extra weight on returns"

    return f"Strategy: {base_strategy}. Focus: {focus}{risk_note}."


def generate_recommendation(
    goal: str,
    risk_profile: str,
    df_features: pd.DataFrame,
    df_norm: pd.DataFrame,
    top_n: int = 5
) -> Dict:
    """Generate stock recommendation based on goal and risk profile."""
    weights = get_goal_weights(goal)
    weights = adjust_weights_for_risk(weights, risk_profile)

    df_scored = score_stocks(df_norm, weights)
    top_stocks = df_scored.head(top_n)

    allocations = get_smooth_allocations(top_stocks["final_score"])

    risk_levels = {}
    stock_insights = {}
    for stock in top_stocks.index.tolist():
        vol = df_features.loc[stock, "volatility"]
        risk_levels[stock] = classify_risk(vol)
        stock_insights[stock] = generate_stock_insight(stock, df_features)

    strategy_text = get_strategy_explanation(goal, risk_profile)

    return {
        "recommended_stocks": top_stocks.index.tolist(),
        "allocations": allocations,
        "scores": top_stocks["final_score"].to_dict(),
        "risk_levels": risk_levels,
        "stock_insights": stock_insights,
        "strategy": strategy_text,
        "weights_used": weights
    }


# ============================================
# RUN RECOMMENDATION ENGINE
# ============================================

print("=" * 60)
print("6.6 RECOMMENDATION ENGINE")
print("=" * 60)

df_features = build_stock_features_dataframe(all_stocks_data)
print(f"\nBuilt features for {len(df_features)} stocks")

df_norm = normalize_features(df_features)
print("Normalized features")

print("TOP STOCKS BY METRIC")
top_return = df_features.nlargest(5, "annual_return")
print("\nHighest Return:")
for stock, row in top_return.iterrows():
    print(f"  {stock}: {row['annual_return']:.1f}%")

top_stable = df_features.nsmallest(5, "volatility")
print("\nLowest Volatility (Most Stable):")
for stock, row in top_stable.iterrows():
    print(f"  {stock}: {row['volatility']:.1f}%")

best_drawdown = df_features.nlargest(5, "max_drawdown")
print("\nBest Drawdown (Crisis Resilient):")
for stock, row in best_drawdown.iterrows():
    print(f"  {stock}: {row['max_drawdown']*100:.1f}% max loss")

all_stocks_data["_feature_analysis"] = df_features
all_stocks_data["_normalized_features"] = df_norm
```

```python
# ============================================
# 6.7 ELEVATED FEATURES
# ============================================
# Enhancement: Dividend yield integration, sector diversification, benchmark comparison

from typing import Dict, Tuple
import numpy as np
import pandas as pd


def compute_dividend_yield(dividend_data: Dict, stock: str, current_price: float = None) -> float:
    """
    Calculate average dividend yield for a stock.

    Args:
        dividend_data: Dictionary of dividend per share by year
        stock: Stock ticker
        current_price: Current price (if None, uses estimated 5% of DPS)

    Returns:
        Average dividend yield in percentage
    """
    if stock not in dividend_data:
        return 0.0

    div_values = list(dividend_data[stock].values())
    if not div_values:
        return 0.0

    avg_dividend = sum(div_values) / len(div_values)

    if current_price and current_price > 0:
        return (avg_dividend / current_price) * 100

    # Fallback: assume dividend yield ~ 3-5% of DPS
    return avg_dividend * 0.04


def enhance_features_with_dividend(
    df_features: pd.DataFrame,
    dividend_data: Dict,
    all_stocks_data: Dict
) -> pd.DataFrame:
    """
    Add dividend yield feature to dataframe.
    """
    df_enhanced = df_features.copy()
    df_enhanced["dividend_yield"] = 0.0

    for stock in df_enhanced.index:
        if stock not in dividend_data:
            continue

        # Get current price from all_stocks_data
        current_price = None
        if stock in all_stocks_data:
            df = all_stocks_data[stock]
            if isinstance(df, pd.DataFrame) and 'close' in df.columns:
                current_price = df['close'].iloc[-1]

        div_yield = compute_dividend_yield(dividend_data, stock, current_price)
        df_enhanced.loc[stock, "dividend_yield"] = div_yield

    return df_enhanced


def get_goal_weights_with_dividend(goal: str) -> Dict[str, float]:
    """
    Return weights including dividend for KPR goals.
    """
    if "Wedding" in goal:
        return {
            "return": 0.30,
            "volatility": 0.40,
            "drawdown": 0.20,
            "dividend": 0.10
        }
    elif "KPR" in goal:
        return {
            "return": 0.30,
            "volatility": 0.25,
            "drawdown": 0.15,
            "dividend": 0.30
        }
    else:  # Education
        return {
            "return": 0.50,
            "volatility": 0.25,
            "drawdown": 0.15,
            "dividend": 0.10
        }


def diversify_by_sector(
    top_stocks: pd.DataFrame,
    sector_map: Dict,
    max_per_sector: int = 2
) -> pd.DataFrame:
    """
    Ensure portfolio has sector diversification.
    Returns filtered stocks maintaining order.
    """
    selected = []
    sector_count = {}

    for stock in top_stocks.index:
        sector = sector_map.get(stock, "Other")

        if sector_count.get(sector, 0) < max_per_sector:
            selected.append(stock)
            sector_count[sector] = sector_count.get(sector, 0) + 1

        if len(selected) >= 5:
            break

    # If no stocks selected (should not happen), return original top 5
    if not selected:
        return top_stocks.head(5)

    return top_stocks.loc[selected]


def calculate_benchmark_return(all_stocks_data: Dict) -> Tuple[float, float, float]:
    """
    Calculate LQ45 benchmark metrics.

    Returns:
        (avg_return, avg_volatility, sharpe)
    """
    returns = []
    volatilities = []

    for stock, df in all_stocks_data.items():
        if not isinstance(df, pd.DataFrame):
            continue
        if 'close' not in df.columns:
            continue

        daily_ret = df["close"].pct_change()
        if len(daily_ret.dropna()) < 100:
            continue

        annual_return = daily_ret.mean() * 252 * 100
        volatility = daily_ret.std() * np.sqrt(252) * 100

        if not np.isnan(annual_return) and not np.isnan(volatility):
            returns.append(annual_return)
            volatilities.append(volatility)

    avg_return = np.mean(returns) if returns else 10.0
    avg_volatility = np.mean(volatilities) if volatilities else 25.0
    sharpe = avg_return / avg_volatility if avg_volatility > 0 else 0.4

    return avg_return, avg_volatility, sharpe


def enhance_recommendation(
    recommendation: Dict,
    df_features: pd.DataFrame,
    all_stocks_data: Dict,
    sector_map: Dict
) -> Dict:
    """
    Add elevated features to existing recommendation.
    """
    enhanced = recommendation.copy()

    # Add sector info
    stock_sectors = {}
    for stock in recommendation["recommended_stocks"]:
        stock_sectors[stock] = sector_map.get(stock, "Other")
    enhanced["sectors"] = stock_sectors

    # Add dividend yield
    dividend_yields = {}
    for stock in recommendation["recommended_stocks"]:
        if "dividend_yield" in df_features.columns:
            dividend_yields[stock] = round(df_features.loc[stock, "dividend_yield"], 2)
        else:
            dividend_yields[stock] = 0.0
    enhanced["dividend_yields"] = dividend_yields

    # Calculate portfolio metrics
    portfolio_return = 0
    portfolio_vol = 0
    portfolio_div = 0

    for stock, allocation in recommendation["allocations"].items():
        weight = allocation / 100
        portfolio_return += df_features.loc[stock, "annual_return"] * weight
        portfolio_vol += df_features.loc[stock, "volatility"] * weight
        if "dividend_yield" in df_features.columns:
            portfolio_div += df_features.loc[stock, "dividend_yield"] * weight

    enhanced["portfolio_return"] = round(portfolio_return, 1)
    enhanced["portfolio_volatility"] = round(portfolio_vol, 1)
    enhanced["portfolio_dividend"] = round(portfolio_div, 2)

    # Benchmark comparison
    benchmark_return, benchmark_vol, benchmark_sharpe = calculate_benchmark_return(all_stocks_data)
    enhanced["benchmark_return"] = round(benchmark_return, 1)
    enhanced["benchmark_volatility"] = round(benchmark_vol, 1)
    enhanced["outperformance"] = round(portfolio_return - benchmark_return, 1)

    # Risk-adjusted comparison
    portfolio_sharpe = portfolio_return / portfolio_vol if portfolio_vol > 0 else 0
    enhanced["portfolio_sharpe"] = round(portfolio_sharpe, 2)
    enhanced["benchmark_sharpe"] = round(benchmark_sharpe, 2)

    # Generate explanation texts
    if enhanced["outperformance"] > 0:
        enhanced["performance_text"] = f"Portfolio expected to outperform benchmark by {enhanced['outperformance']:.1f}%"
    else:
        enhanced["performance_text"] = f"Portfolio aligned with LQ45 benchmark (within {abs(enhanced['outperformance']):.1f}%)"

    if "KPR" in recommendation.get("strategy", ""):
        enhanced["dividend_text"] = f"Estimated annual dividend: {enhanced['portfolio_dividend']:.1f}% yield"
    else:
        enhanced["dividend_text"] = ""

    return enhanced


# ============================================
# RUN ELEVATED FEATURES (Optional)
# ============================================

print("=" * 60)
print("6.7 ELEVATED FEATURES")
print("=" * 60)

# Add dividend to features if dividend data available
if 'dividend_data_of_36' in dir():
    df_features_enhanced = enhance_features_with_dividend(
        df_features, dividend_data_of_36, all_stocks_data
    )
    print(f"\nEnhanced features with dividend data")
    print(f"Top 5 dividend yields:")
    top_div = df_features_enhanced.nlargest(5, "dividend_yield")
    for stock, row in top_div.iterrows():
        print(f"  {stock}: {row['dividend_yield']:.2f}%")
else:
    print("\nNo dividend data available. Skipping dividend enhancement.")

# Benchmark analysis
benchmark_return, benchmark_vol, benchmark_sharpe = calculate_benchmark_return(all_stocks_data)
print(f"\nLQ45 BENCHMARK METRICS:")
print(f"  Average Return: {benchmark_return:.1f}%")
print(f"  Average Volatility: {benchmark_vol:.1f}%")
print(f"  Sharpe Ratio: {benchmark_sharpe:.2f}")

print("\n" + "=" * 60)
print("Elevated Features Ready")
print("=" * 60)

# Store for later use
all_stocks_data["_elevated_features"] = df_features_enhanced if 'df_features_enhanced' in dir() else df_features
all_stocks_data["_benchmark"] = {
    "return": benchmark_return,
    "volatility": benchmark_vol,
    "sharpe": benchmark_sharpe}
```

```python
# ============================================
# 6.8 PERSONALIZED INPUT
# ============================================
# Dynamic weight adjustment based on user's timeline and risk capacity

from typing import Dict


def adjust_weights_by_timeline(
    weights: Dict[str, float],
    years: float,
    goal: str
) -> Dict[str, float]:
    """
    Adjust weights based on how close the deadline is.

    - Short timeline (< 2 years): prioritize stability more
    - Long timeline (> 10 years): prioritize growth more
    """
    adjusted = weights.copy()

    if "Wedding" in goal:
        if years < 1.5:
            # Very short timeline (1-1.5 years) -> extreme caution
            adjusted["volatility"] = adjusted.get("volatility", 0) + 0.15
            adjusted["return"] = adjusted.get("return", 0) - 0.15
        elif years > 2.5:
            # Longer wedding timeline -> can take slightly more risk
            adjusted["return"] = adjusted.get("return", 0) + 0.05
            adjusted["volatility"] = adjusted.get("volatility", 0) - 0.05

    elif "KPR" in goal:
        if years < 3.5:
            # Short KPR timeline
            adjusted["volatility"] = adjusted.get("volatility", 0) + 0.10
            adjusted["dividend"] = adjusted.get("dividend", 0) + 0.05
            adjusted["return"] = adjusted.get("return", 0) - 0.15
        elif years > 4.5:
            # Longer KPR timeline
            adjusted["return"] = adjusted.get("return", 0) + 0.10
            adjusted["dividend"] = adjusted.get("dividend", 0) - 0.05

    else:  # Education
        if years < 12:
            # Closer to college (10-12 years) -> reduce risk gradually
            adjusted["volatility"] = adjusted.get("volatility", 0) + 0.05
            adjusted["return"] = adjusted.get("return", 0) - 0.05
        elif years > 15:
            # Very long horizon -> higher growth focus
            adjusted["return"] = adjusted.get("return", 0) + 0.10
            adjusted["volatility"] = adjusted.get("volatility", 0) - 0.10

    # Remove negative values
    for k in adjusted:
        adjusted[k] = max(0, adjusted[k])

    # Re-normalize to sum to 1
    total = sum(adjusted.values())
    if total > 0:
        return {k: v / total for k, v in adjusted.items()}
    return weights


def adjust_weights_by_risk_capacity(
    weights: Dict[str, float],
    risk_capacity: str
) -> Dict[str, float]:
    """
    Adjust weights based on user's loss tolerance.

    risk_capacity:
        "Low": Can tolerate <10% loss
        "Medium": Can tolerate 10-25% loss
        "High": Can tolerate >25% loss
    """
    adjusted = weights.copy()

    if risk_capacity == "Low":
        adjusted["volatility"] = adjusted.get("volatility", 0) + 0.20
        adjusted["drawdown"] = adjusted.get("drawdown", 0) + 0.15
        adjusted["return"] = adjusted.get("return", 0) - 0.35

    elif risk_capacity == "Medium":
        # Balanced, no extra adjustment
        pass

    elif risk_capacity == "High":
        adjusted["return"] = adjusted.get("return", 0) + 0.20
        adjusted["volatility"] = adjusted.get("volatility", 0) - 0.15
        adjusted["drawdown"] = adjusted.get("drawdown", 0) - 0.05

    # Remove negative values
    for k in adjusted:
        adjusted[k] = max(0, adjusted[k])

    # Re-normalize
    total = sum(adjusted.values())
    if total > 0:
        return {k: v / total for k, v in adjusted.items()}
    return weights


def adjust_weights_by_liquidity(
    weights: Dict[str, float],
    need_liquidity: bool
) -> Dict[str, float]:
    """
    Adjust weights if user needs quick access to funds.
    """
    if not need_liquidity:
        return weights

    adjusted = weights.copy()

    adjusted["volatility"] = adjusted.get("volatility", 0) + 0.10
    adjusted["dividend"] = adjusted.get("dividend", 0) + 0.10
    adjusted["return"] = adjusted.get("return", 0) - 0.15
    adjusted["drawdown"] = adjusted.get("drawdown", 0) + 0.05

    # Remove negative values
    for k in adjusted:
        adjusted[k] = max(0, adjusted[k])

    # Re-normalize
    total = sum(adjusted.values())
    if total > 0:
        return {k: v / total for k, v in adjusted.items()}
    return weights


def get_personalized_weights(
    goal: str,
    years: float,
    risk_capacity: str,
    need_liquidity: bool,
    base_weights: Dict[str, float]
) -> Dict[str, float]:
    """
    Combine all personalization adjustments.
    """
    weights = base_weights.copy()

    weights = adjust_weights_by_timeline(weights, years, goal)
    weights = adjust_weights_by_risk_capacity(weights, risk_capacity)
    weights = adjust_weights_by_liquidity(weights, need_liquidity)

    return weights


def generate_personalized_recommendation(
    goal: str,
    risk_profile: str,
    years: float,
    risk_capacity: str,
    need_liquidity: bool,
    top_n: int = 5
) -> Dict:
    """
    Generate personalized stock recommendation.
    """
    # Get base weights based on goal (using existing functions)
    if "Wedding" in goal:
        base_weights = {"return": 0.30, "volatility": 0.45, "drawdown": 0.25, "dividend": 0.00}
    elif "KPR" in goal:
        base_weights = {"return": 0.35, "volatility": 0.35, "drawdown": 0.20, "dividend": 0.10}
    else:
        base_weights = {"return": 0.55, "volatility": 0.25, "drawdown": 0.20, "dividend": 0.00}

    # Adjust for risk profile
    if risk_profile == "Conservative":
        base_weights["volatility"] += 0.10
        base_weights["drawdown"] += 0.10
        base_weights["return"] -= 0.20
    elif risk_profile == "Aggressive":
        base_weights["return"] += 0.15
        base_weights["volatility"] -= 0.10
        base_weights["drawdown"] -= 0.05

    # Remove negative and re-normalize base weights
    for k in base_weights:
        base_weights[k] = max(0, base_weights[k])
    total = sum(base_weights.values())
    if total > 0:
        base_weights = {k: v / total for k, v in base_weights.items()}

    # Apply personalization
    personalized_weights = get_personalized_weights(
        goal, years, risk_capacity, need_liquidity, base_weights
    )

    # Use existing df_features and df_norm from 6.6
    # Score stocks using the weights
    df_score = df_norm.copy()
    df_score["final_score"] = 0

    if "return" in personalized_weights:
        df_score["final_score"] += personalized_weights["return"] * df_score["return_norm"]
    if "volatility" in personalized_weights:
        df_score["final_score"] += personalized_weights["volatility"] * df_score["vol_norm"]
    if "drawdown" in personalized_weights:
        df_score["final_score"] += personalized_weights["drawdown"] * df_score["drawdown_norm"]
    if "dividend" in personalized_weights and "dividend_norm" in df_score.columns:
        df_score["final_score"] += personalized_weights["dividend"] * df_score["dividend_norm"]

    df_scored = df_score.sort_values("final_score", ascending=False)
    top_stocks = df_scored.head(top_n)

    # Simple allocation
    allocations = {stock: round(100 / top_n, 1) for stock in top_stocks.index}
    remainder = 100 - sum(allocations.values())
    if remainder != 0:
        first_stock = list(allocations.keys())[0]
        allocations[first_stock] += remainder

    # Personalization note
    personalization_note = []
    if years < 2 and "Wedding" in goal:
        personalization_note.append("Short timeline: portfolio weighted toward stability")
    if years > 12 and "Education" in goal:
        personalization_note.append("Long timeline: higher growth allocation")
    if risk_capacity == "Low":
        personalization_note.append("Low loss tolerance: prioritized capital protection")
    if need_liquidity:
        personalization_note.append("Liquidity need: included dividend-paying stocks")

    return {
        "recommended_stocks": top_stocks.index.tolist(),
        "allocations": allocations,
        "scores": top_stocks["final_score"].to_dict(),
        "weights_used": personalized_weights,
        "personalization_note": " | ".join(personalization_note) if personalization_note else "Standard profile applied"
    }


# ============================================
# DEMO
# ============================================

print("=" * 60)
print("6.8 PERSONALIZED INPUT")
print("=" * 60)

print("\nExample: Different inputs produce different recommendations")

# Test scenarios
test_cases = [
    {"goal": "Wedding Fund (1-3 years)", "risk_profile": "Conservative", "years": 1.5, "risk_capacity": "Low", "need_liquidity": True},
    {"goal": "Wedding Fund (1-3 years)", "risk_profile": "Conservative", "years": 2.5, "risk_capacity": "Medium", "need_liquidity": False},
    {"goal": "Child Education (10-18 years)", "risk_profile": "Aggressive", "years": 15, "risk_capacity": "High", "need_liquidity": False},
]

for i, case in enumerate(test_cases, 1):
    print(f"\n--- Scenario {i} ---")
    print(f"Goal: {case['goal']}")
    print(f"Years: {case['years']}, Risk Capacity: {case['risk_capacity']}, Liquidity Need: {case['need_liquidity']}")

    rec = generate_personalized_recommendation(
        goal=case['goal'],
        risk_profile=case['risk_profile'],
        years=case['years'],
        risk_capacity=case['risk_capacity'],
        need_liquidity=case['need_liquidity'],
        top_n=4
    )
    print(f"Recommended: {rec['recommended_stocks']}")
    print(f"Note: {rec['personalization_note']}")

all_stocks_data["_personalized_weights"] = personalized_weights if 'personalized_weights' in dir() else {}
```

```python
# ============================================
# 6.9 PORTFOLIO MANAGEMENT & SHIFT PLANNING
# ============================================
# Allows users to record their current stock positions
# Calculates unrealized capital gain/loss
# Provides shift plan to recommended portfolio

from ipywidgets import interact, IntText, FloatSlider, Dropdown, DatePicker, Output, Button, VBox, HBox
from IPython.display import display, clear_output
from datetime import date, timedelta
import pandas as pd
import numpy as np


class PortfolioManager:
    """
    Manages user's existing stock positions and provides shift planning.
    """

    def __init__(self, all_stocks_data):
        self.all_stocks_data = all_stocks_data
        self.user_positions = []

    def get_price_on_date(self, stock: str, target_date: date) -> float:
        """Get historical price of a stock on a specific trading day."""
        df = self.all_stocks_data.get(stock)
        if df is None or 'close' not in df.columns:
            return None

        # Find closest available trading day (cannot exceed target_date)
        available_dates = df[df.index <= pd.to_datetime(target_date)].index
        if len(available_dates) == 0:
            return None

        closest_date = available_dates[-1]
        return df.loc[closest_date, 'close']

    def add_position(self, stock: str, purchase_date: date, purchase_price: float, lot_size: int):
        """
        Add a stock position.
        Note: 1 lot = 100 shares in IDX
        """
        # Validate purchase price if not provided
        if purchase_price == 0:
            purchase_price = self.get_price_on_date(stock, purchase_date)
            if purchase_price is None:
                print(f"Warning: Could not determine price for {stock} on {purchase_date}")
                return

        shares = lot_size * 100
        position = {
            'stock': stock,
            'purchase_date': purchase_date,
            'purchase_price': purchase_price,
            'lot_size': lot_size,
            'shares': shares,
            'cost_basis': purchase_price * shares
        }
        self.user_positions.append(position)

    def calculate_current_value(self, stock: str) -> float:
        """Get current market price of a stock."""
        df = self.all_stocks_data.get(stock)
        if df is None or 'close' not in df.columns:
            return None
        return df['close'].iloc[-1]

    def get_unrealized_gain_loss(self) -> pd.DataFrame:
        """Calculate unrealized capital gain/loss for all positions."""
        results = []
        total_cost = 0
        total_current = 0

        for pos in self.user_positions:
            current_price = self.calculate_current_value(pos['stock'])
            if current_price is None:
                continue

            current_value = current_price * pos['shares']
            gain_loss = current_value - pos['cost_basis']
            gain_loss_pct = (gain_loss / pos['cost_basis']) * 100

            results.append({
                'Stock': pos['stock'],
                'Purchase Date': pos['purchase_date'].strftime('%Y-%m-%d'),
                'Purchase Price': pos['purchase_price'],
                'Current Price': current_price,
                'Lot Size': pos['lot_size'],
                'Shares': pos['shares'],
                'Cost Basis': pos['cost_basis'],
                'Current Value': current_value,
                'Unrealized Gain/Loss': gain_loss,
                'Return %': gain_loss_pct
            })

            total_cost += pos['cost_basis']
            total_current += current_value

        df = pd.DataFrame(results)
        return df, total_cost, total_current

    def generate_shift_plan(self, recommended_stocks: list, recommended_allocations: dict) -> dict:
        """
        Generate plan to shift from current positions to recommended portfolio.
        """
        current_df, total_cost, total_current = self.get_unrealized_gain_loss()

        plan = {
            'current_portfolio': current_df.to_dict('records'),
            'total_invested': total_cost,
            'current_value': total_current,
            'total_return_pct': ((total_current - total_cost) / total_cost * 100) if total_cost > 0 else 0,
            'sell_recommendations': [],
            'buy_recommendations': [],
            'notes': []
        }

        # Overlap analysis: stocks that are in both current and recommended
        current_stocks = set([p['Stock'] for p in plan['current_portfolio']])
        recommended_stocks_set = set(recommended_stocks)

        overlap = current_stocks.intersection(recommended_stocks_set)
        to_sell = current_stocks - recommended_stocks_set
        to_buy = recommended_stocks_set - current_stocks

        # Generate sell recommendations
        for pos in plan['current_portfolio']:
            if pos['Stock'] in to_sell:
                plan['sell_recommendations'].append({
                    'stock': pos['Stock'],
                    'shares': pos['Shares'],
                    'current_price': pos['Current Price'],
                    'estimated_proceeds': pos['Current Value'],
                    'reason': f"Not in recommended portfolio for your goal and risk profile"
                })
            elif pos['Stock'] in overlap:
                # Keep but may need to adjust allocation
                plan['notes'].append(f"Stock {pos['Stock']} is already in your portfolio and recommended. Maintain position.")

        # Generate buy recommendations
        target_portfolio_value = total_current  # Maintain same capital
        recommended_allocation = {stock: recommended_allocations.get(stock, 0) / 100 for stock in recommended_stocks}

        for stock in to_buy:
            target_value = target_portfolio_value * recommended_allocation.get(stock, 0)
            shares_to_buy = int(target_value / self.calculate_current_value(stock))
            if shares_to_buy > 0:
                plan['buy_recommendations'].append({
                    'stock': stock,
                    'target_allocation_pct': recommended_allocation.get(stock, 0) * 100,
                    'estimated_shares': shares_to_buy,
                    'estimated_cost': shares_to_buy * self.calculate_current_value(stock),
                    'reason': f"Recommended for your {recommended_allocation.get(stock, 0)*100:.0f}% allocation"
                })

        return plan


# ============================================
# LQ45 OVERVIEW (Educational Component)
# ============================================

LQ45_OVERVIEW = """
The LQ45 index consists of 45 highly liquid stocks listed on the Indonesia Stock Exchange (IDX).
It is rebalanced every six months (February and August) based on liquidity and market capitalization.

Why LQ45 as a benchmark for financial planning:
- Liquidity: LQ45 stocks are the most actively traded, ensuring you can enter and exit positions with minimal price impact
- Transparency: All constituents are publicly traded with regular financial disclosures
- Track record: The index has demonstrated resilience through multiple market cycles including the 2008 GFC and 2020 COVID crisis
- Diversification: Covers 9 sectors (Banking, Consumer, Energy, Mining, Property, Telecom, Automotive, Infrastructure, Basic Materials)

Current Market Context (as of February 2025):
- The index is recovering from the post-pandemic volatility period
- Banking and Consumer sectors have shown relative strength
- Energy stocks have experienced higher volatility due to global commodity price fluctuations
"""


# ============================================
# INTERACTIVE PORTFOLIO INPUT
# ============================================

portfolio_output = Output()

def create_portfolio_input_ui(portfolio_manager):
    """Create UI for users to input their existing stock positions."""

    print("\n" + "=" * 60)
    print("YOUR CURRENT PORTFOLIO")
    print("=" * 60)
    print(LQ45_OVERVIEW)
    print("\nPlease enter your existing stock positions below.")
    print("If you have no existing positions, you can skip this section.\n")

    # Input fields for a single position
    stock = Dropdown(options=emiten_clean, description='Stock Ticker:')
    has_position = st.radio("Do you own this stock?", ["No", "Yes"])

    if has_position == "Yes":
        purchase_date = DatePicker(description='Purchase Date:', value=date(2020, 1, 1))
        purchase_price = IntText(description='Purchase Price (Rp):', value=0, placeholder='Leave 0 to auto-fill from date')
        lot_size = IntText(description='Lot Size (1 lot = 100 shares):', value=1)

        if st.button("Add Position", key="add_position_btn"):
            if purchase_price.value == 0:
                price = portfolio_manager.get_price_on_date(stock.value, purchase_date.value)
                if price:
                    purchase_price.value = price
                else:
                    st.warning(f"Could not auto-fill price for {stock.value} on {purchase_date.value}")

            portfolio_manager.add_position(stock.value, purchase_date.value, purchase_price.value, lot_size.value)
            st.success(f"Added {stock.value}: {lot_size.value} lot at Rp {purchase_price.value:,.0f}")
            st.rerun()

    if len(portfolio_manager.user_positions) > 0:
        st.markdown("### Current Positions")
        df, _, _ = portfolio_manager.get_unrealized_gain_loss()
        st.dataframe(df[['Stock', 'Purchase Date', 'Purchase Price', 'Current Price', 'Lot Size', 'Return %']])

        if st.button("Clear All Positions", key="clear_positions"):
            portfolio_manager.user_positions = []
            st.rerun()


# ============================================
# SHIFT PLAN EXECUTION
# ============================================

def generate_shift_plan_ui(portfolio_manager, recommended_stocks, recommended_allocations):
    """Generate and display shift plan."""

    plan = portfolio_manager.generate_shift_plan(recommended_stocks, recommended_allocations)

    print("\n" + "=" * 60)
    print("PORTFOLIO SHIFT PLAN")
    print("=" * 60)

    print(f"\nCurrent Portfolio Summary:")
    print(f"  Total Invested: Rp {plan['total_invested']:,.0f}")
    print(f"  Current Market Value: Rp {plan['current_value']:,.0f}")
    print(f"  Unrealized Return: {plan['total_return_pct']:.1f}%")

    if plan['sell_recommendations']:
        print("\n" + "-" * 40)
        print("SELL RECOMMENDATIONS")
        print("-" * 40)
        for sell in plan['sell_recommendations']:
            print(f"\n  Stock: {sell['stock']}")
            print(f"    Current Price: Rp {sell['current_price']:,.0f}")
            print(f"    Shares to Sell: {sell['shares']:,}")
            print(f"    Estimated Proceeds: Rp {sell['estimated_proceeds']:,.0f}")
            print(f"    Reason: {sell['reason']}")

    if plan['buy_recommendations']:
        print("\n" + "-" * 40)
        print("BUY RECOMMENDATIONS")
        print("-" * 40)
        for buy in plan['buy_recommendations']:
            print(f"\n  Stock: {buy['stock']}")
            print(f"    Target Allocation: {buy['target_allocation_pct']:.0f}%")
            print(f"    Estimated Shares: {buy['estimated_shares']:,} shares")
            print(f"    Estimated Cost: Rp {buy['estimated_cost']:,.0f}")
            print(f"    Reason: {buy['reason']}")

    if plan['notes']:
        print("\n" + "-" * 40)
        print("ADDITIONAL NOTES")
        print("-" * 40)
        for note in plan['notes']:
            print(f"  {note}")

    print("\n" + "=" * 60)
    print("EXECUTION GUIDANCE")
    print("=" * 60)
    print("""
    1. For sell recommendations: Consider selling gradually if the position has significant unrealized loss
    2. For buy recommendations: Use limit orders to avoid paying excessive premiums
    3. Tax Consideration: Capital gains tax on IDX listed stocks is 0.1% of proceeds
    4. Transaction Cost: Broker fees typically range from 0.1% to 0.3% per transaction
    """)

def run_shift_planning(portfolio_manager, recommended_stocks, recommended_allocations):
    with portfolio_output:
        clear_output(wait=True)
        generate_shift_plan_ui(portfolio_manager, recommended_stocks, recommended_allocations)
```
